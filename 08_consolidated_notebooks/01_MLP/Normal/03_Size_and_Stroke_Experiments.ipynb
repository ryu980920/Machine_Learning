## 사용법
원본 코드 셀은 `%%archive_source`로 등록하는 실제 원본 코드이며, 함수명 충돌을 막기 위해 현재 커널에서 임의 실행하지 않는다. 맨 아래의 실행 셀에서 원본 스크립트/Notebook을 고정 커밋의 **별도 checkout**에서 실행한다. 각 실험의 실행 인자·입력·사전 조건은 위 원본 로그/코드로 확인한다. 기존 로그·결과와 새 출력이 섞이지 않도록 독립 복사본을 사용한다. Original Test/교수님 Test를 포함한 파일은 실행 환경에 두지 않는다. 과거 스크립트가 해당 자료를 요구할 경우 실행하지 않는다. 자동 재실행 및 성능 동등성은 미검증.

원본 실행을 위한 로그 디렉터리를 별도로 지정한다.

```python
# 공통 실행 제어 — Run All은 원본 코드 등록만 하며 학습하지 않는다.
from pathlib import Path
import hashlib, json, subprocess, sys, datetime
from IPython import get_ipython
SOURCE_CELLS = {}
def archive_source(line, cell):
    SOURCE_CELLS[line.strip()] = cell
get_ipython().register_magic_function(archive_source, magic_kind="cell", magic_name="archive_source")
# 실행 시에만 설정: 기존 작업 트리가 아닌 임시 checkout (git worktree add --detach ... <ref>)
WORKTREE = None
LOG_DIR = None
RUN_PLAN = []  # 예: [("원본/스크립트.py", ["--인자", "값"])]
ALLOW_EXECUTION = False

def blob_sha(data):
    return hashlib.sha1(f"blob {len(data)}\0".encode()+data).hexdigest()

def get_verified_source(path):
    if WORKTREE is None:
        raise ValueError("해당 커밋의 별도 checkout 경로 WORKTREE를 지정하세요.")
    root=Path(WORKTREE).expanduser().resolve()
    entry=SOURCE_INDEX[path]
    head=subprocess.check_output(["git","-C",str(root),"rev-parse","HEAD"],text=True).strip()
    if head != entry["ref"]:
        raise ValueError(f"checkout ref 불일치: {head} != {entry['ref']}")
    p=root/path
    if blob_sha(p.read_bytes()) != entry["sha"]:
        raise ValueError(f"원본 Git blob 불일치: {path}")
    if path in SOURCE_CELLS:
        if p.read_text(encoding="utf-8-sig").rstrip("\n") != SOURCE_CELLS[path].rstrip("\n").lstrip("\ufeff"):
            raise ValueError(f"Notebook 보존 코드와 원본이 다릅니다: {path}")
    return root,p

def show_original(path):
    from IPython.display import display, Code
    _,p=get_verified_source(path)
    display(Code(p.read_text(encoding="utf-8"),language="python" if p.suffix==".py" else "json"))

def execute_original(path, args):
    if not ALLOW_EXECUTION:
        raise RuntimeError("명시적 실행 승인이 필요합니다.")
    root,p=get_verified_source(path)
    if LOG_DIR is None:
        raise ValueError("새 외부 LOG_DIR를 지정하세요.")
    logs=Path(LOG_DIR).expanduser().resolve();logs.mkdir(parents=True,exist_ok=True)
    if logs==root or root in logs.parents:
        raise ValueError("LOG_DIR는 checkout 밖에 두세요.")
    stamp=datetime.datetime.now(datetime.timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
    if p.suffix==".ipynb":
        cmd=[sys.executable,"-m","jupyter","nbconvert","--to","notebook","--execute",str(p),"--output-dir",str(logs),"--output",f"{p.stem}_{stamp}.ipynb",*args]
    else:
        cmd=[sys.executable,str(p),*args]
    log=logs/f"{p.stem}_{stamp}.log"
    with log.open("x",encoding="utf-8") as f:
        f.write(json.dumps({"source":path,**SOURCE_INDEX[path],"command":cmd,"cwd":str(p.parent)},ensure_ascii=False)+"\n");f.flush()
        result=subprocess.run(cmd,cwd=p.parent,stdout=f,stderr=subprocess.STDOUT)
        f.write(f"\nreturncode={result.returncode}\n")
    print(log)
    result.check_returncode()
```

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp007_size_com_stroke_followup.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""Normal HypB EXP007, controlled Stroke addition after Size -> CoM.

Official train 76,800 only for BOTH MLP and fitted stroke normalizer.
Developer evaluations: Official Validation 19,200; existing Hand-made 2,560.
No official Original Test, professor Test or Hand-made learning.
"""
from __future__ import annotations
import argparse
import csv
import gzip
import hashlib
import json
import platform
import time
import warnings
from pathlib import Path

import numpy as np
import scipy
import sklearn
from joblib import dump
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix

from exp006_size_com_comparison import (
    ROOT, WRITERS, LABELS, PAIRS, load_official, load_writers,
    PixelScaler255, image_geometry, process, center_batch,
    score, evaluate_group, save_csv, sha256,
)
from stroke_normalization import StrokeWidthNormalizer

CONDITIONS = ("size_com", "size_com_stroke030", "size_com_stroke060")
ALPHA = {"size_com": None, "size_com_stroke030": 0.30,
         "size_com_stroke060": 0.60}


def train(condition, seed, out):
    out = Path(out)
    if out.exists():
        raise RuntimeError("Refusing to overwrite existing result: " + str(out))
    out.mkdir(parents=True)
    official = load_official(ROOT / "02_data" / "datasets" /
                             "MNIST_EMNIST_official_split_112000.npz")
    handmade = load_writers(ROOT / "02_data" / "Hand_made")
    target = int(np.median(image_geometry(official["X_train"])["long_side"]))
    if target != 20:
        raise AssertionError("Expected official Train BBox long-side median=20")

    raw_hm = np.concatenate([handmade[w]["X"] for w in WRITERS])
    label_hm = np.concatenate([handmade[w]["y"] for w in WRITERS])
    writers = np.repeat(np.array(WRITERS, dtype=int), 640)

    # All conditions share identical Combined-derived Size -> CoM input.
    x_train = process(official["X_train"], "size_com", target)
    x_val = process(official["X_val"], "size_com", target)
    x_hm = process(raw_hm, "size_com", target)
    stroke_fit = None
    alpha = ALPHA[condition]
    if alpha is not None:
        stroke_fit = StrokeWidthNormalizer(strength=alpha, threshold=64,
            fit_quantiles=(0.2, 0.8), preserve_topology=True,
            apply_center=False).fit(x_train)
        def apply_stroke_then_center(x):
            x = stroke_fit.transform(x)
            return center_batch(x, "mass").astype(np.float32)
        x_train = apply_stroke_then_center(x_train)
        x_val = apply_stroke_then_center(x_val)
        x_hm = apply_stroke_then_center(x_hm)

    assert x_train.shape == (76800, 784)
    assert x_val.shape == (19200, 784)
    assert x_hm.shape == (2560, 784)
    assert np.isfinite(x_train).all() and x_train.min() >= 0 and x_train.max() <= 255

    model = Pipeline([("scale_0_1", PixelScaler255()),
                      ("model", MLPClassifier(random_state=seed))])
    start = time.monotonic()
    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")
        model.fit(x_train, official["y_train"])
    duration = time.monotonic() - start
    val_pred = model.predict(x_val).astype(int)
    hm_pred = model.predict(x_hm).astype(int)
    official_scores = evaluate_group(official["y_val"], val_pred)
    handmade_scores = evaluate_group(label_hm, hm_pred, writers)

    cm = confusion_matrix(label_hm, hm_pred, labels=np.arange(16))
    save_csv(out / "pair_errors.csv",
        [{"actual":a, "predicted":b,
          "count":int(cm[LABELS.index(a),LABELS.index(b)])} for a,b in PAIRS])
    save_csv(out / "confusion_matrix.csv",
        [{"actual":LABELS[i],
          **{LABELS[j]:int(cm[i,j]) for j in range(16)}} for i in range(16)])
    with gzip.open(out / "handmade_predictions.csv.gz", "wt",
                   encoding="utf-8", newline="") as f:
        writer=csv.writer(f)
        writer.writerow(["writer","index","true","pred","correct"])
        for k in range(2560):
            writer.writerow([int(writers[k]), k % 640,
                LABELS[int(label_hm[k])],LABELS[int(hm_pred[k])],
                int(label_hm[k] == hm_pred[k])])
    save_csv(out / "loss_curve.csv",
        [{"iteration":i,"loss":float(v)} for i,v in
         enumerate(model.named_steps["model"].loss_curve_, 1)])
    dump(model, out/"mlp.joblib", compress=3)
    before_geo = image_geometry(process(raw_hm, "size_com", target))
    after_geo = image_geometry(x_hm)
    manifest = {
        "experiment":"Normal-HypB-EXP007", "status":"trained",
        "condition":condition, "seed":seed, "stroke_strength":alpha,
        "method":"Size -> CoM -> Stroke -> CoM" if alpha is not None else "Size -> CoM",
        "target_bbox_longside_px":target, "bbox_threshold":32,
        "stroke_threshold":64, "stroke_fit_only_official_train":True,
        "stroke_q20":None if stroke_fit is None else float(stroke_fit.q_low_),
        "stroke_q80":None if stroke_fit is None else float(stroke_fit.q_high_),
        "stroke_median":None if stroke_fit is None else float(stroke_fit.median_),
        "stroke_topology_guard":alpha is not None,
        "official_train_n":76800, "official_validation_n":19200, "handmade_dev_n":2560,
        "official_original_test_used":False, "professor_test_used":False,
        "handmade_training_used":False,
        "official_sha256":official["sha256"],
        "writer_sha256":{str(w):handmade[w]["sha256"] for w in WRITERS},
        "training_seconds":float(duration),
        "mlp_iterations":int(model.named_steps["model"].n_iter_),
        "fit_warnings":[f"{v.category.__name__}: {v.message}" for v in caught],
        "official_validation":official_scores,
        "handmade":handmade_scores,
        "handmade_center_offset_px_before_stroke":float(before_geo["com_offset"].mean()),
        "handmade_center_offset_px_after_stroke":float(after_geo["com_offset"].mean()),
        "handmade_mass_change":float(x_hm.sum()-process(raw_hm,"size_com",target).sum()),
        "numpy":np.__version__,"scipy":scipy.__version__,"sklearn":sklearn.__version__,
        "python":platform.python_version(),"model_sha256":sha256(out/"mlp.joblib"),
    }
    (out/"run_manifest.json").write_text(json.dumps(manifest,indent=2,
                        ensure_ascii=False),encoding="utf-8")
    print("EXP007 COMPLETE",json.dumps({
        "condition":condition,"seed":seed,
        "official":official_scores["all"],
        "handmade":handmade_scores},ensure_ascii=False),flush=True)


def aggregate(inputs,out):
    inputs=Path(inputs);out=Path(out)
    if out.exists():
        raise RuntimeError("Refusing to overwrite existing output "+str(out))
    out.mkdir(parents=True)
    files=list(inputs.glob("normal-exp007-*/run_manifest.json"))
    if len(files)!=9:
        raise RuntimeError(f"Expected 9 complete runs, got {len(files)}")
    runs={}; metric_rows=[]
    for p in files:
        j=json.loads(p.read_text(encoding="utf-8"))
        key=(j["condition"],int(j["seed"]))
        if key in runs:raise AssertionError("Duplicate run "+str(key))
        if j["official_original_test_used"] or j["professor_test_used"] or j["handmade_training_used"]:
            raise AssertionError("Data boundary violated")
        runs[key]=(p.parent,j)
        for ds in ("official_validation","handmade"):
            for group,s in j[ds].items():
                metric_rows.append({"condition":key[0],"seed":key[1],
                                    "dataset":ds,"group":group,**s})
    if set(runs)!={(c,s) for c in CONDITIONS for s in (42,43,44)}:
        raise AssertionError("Missing method/seed")
    save_csv(out/"metrics_all.csv",metric_rows)

    def predictions(folder):
        with gzip.open(folder/"handmade_predictions.csv.gz","rt",
                       encoding="utf-8",newline="") as f:
            return {(int(r["writer"]),int(r["index"])):r
                    for r in csv.DictReader(f)}
    paired=[]
    for seed in (42,43,44):
        a=predictions(runs["size_com",seed][0])
        for name in CONDITIONS[1:]:
            b=predictions(runs[name,seed][0])
            if set(a)!=set(b) or len(a)!=2560:
                raise AssertionError("Paired predictions missing")
            for grp in ("all","digits","letters"):
                keys=[k for k in a if grp=="all" or
                      ((a[k]["true"] in "0123456789")==(grp=="digits"))]
                for k in keys:
                    if a[k]["true"]!=b[k]["true"]:
                        raise AssertionError("True labels mismatch")
                corrected=sum(a[k]["correct"]=="0" and b[k]["correct"]=="1" for k in keys)
                induced=sum(a[k]["correct"]=="1" and b[k]["correct"]=="0" for k in keys)
                paired.append({"seed":seed,"candidate":name,"group":grp,
                    "n":len(keys),"corrected":corrected,"new_errors":induced,
                    "net_correct":corrected-induced,
                    "delta_accuracy_pp":100*(corrected-induced)/len(keys)})
    save_csv(out/"paired_vs_size_com.csv",paired)

    lines=["# EXP유007 Size -> CoM -> Stroke -> CoM (review)",
       "", "Only development data were evaluated. Official Test / professor Test unused.",
       "All nine models trained on identical 76,800 official Train rows.",
       "Stroke strength was fixed to 0.30 or 0.60 from earlier HypA experiments.",
       "", "| Condition | Seed | Official Val | Hand-made all | Digits | Letters |",
       "|---|---:|---:|---:|---:|---:|"]
    for c in CONDITIONS:
        for seed in (42,43,44):
            d=runs[c,seed][1]
            lines.append(f"| {c} | {seed} | "
                f"{100*d['official_validation']['all']['accuracy']:.3f}% | "
                f"{100*d['handmade']['all']['accuracy']:.3f}% | "
                f"{100*d['handmade']['digits']['accuracy']:.3f}% | "
                f"{100*d['handmade']['letters']['accuracy']:.3f}% |")
    lines += ["","## Paired accuracy changes relative to Size -> CoM",
       "","| Candidate | Seed | Group | Corrected | Newly wrong | Net |",
       "|---|---:|---|---:|---:|---:|"]
    for d in paired:
        lines.append(f"| {d['candidate']} | {d['seed']} | {d['group']} | "
                     f"{d['corrected']} | {d['new_errors']} | {d['net_correct']} |")
    lines += ["","No automatic final model adoption or GitHub main modification.",
      "New writer independent validation required before generalization claims."]
    (out/"SUMMARY.md").write_text("\n".join(lines)+"\n",encoding="utf-8")
    print("EXP007 AGGREGATION COMPLETE",len(metric_rows), "metric rows",flush=True)


def main():
    p=argparse.ArgumentParser(description=__doc__)
    p.add_argument("--condition",choices=CONDITIONS)
    p.add_argument("--seed",type=int,choices=(42,43,44))
    p.add_argument("--out",type=Path,required=True)
    p.add_argument("--aggregate",action="store_true")
    p.add_argument("--inputs",type=Path)
    a=p.parse_args()
    if a.aggregate:
        if not a.inputs:p.error("--aggregate requires --inputs")
        aggregate(a.inputs,a.out)
    else:
        if a.condition is None or a.seed is None:
            p.error("--condition and --seed required")
        train(a.condition,a.seed,a.out)

if __name__=="__main__":
    main()
```

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp009_c9_augmentation.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""Normal HypB EXP009: training-set augmentation using Original and exact HypA C9.

Comparison preregistered in HypA EXP004 future experiment section:
  A C9-only, B raw-only, C stratified 50:50 replacement,
  D raw+C9, E raw twice.
For equal stochastic-gradient-update budget, A/B/C use 200 epochs on 76,800;
D/E use 100 epochs on 153,600. All have 76,800 nominal minibatches
(batch_size='auto'=200). Train/val is the original source-safe split.
Each model evaluated on both raw and C9 transformed Official Val and Handmade.

No Original Test, no professor Test, no handmade fit or training.
"""
from __future__ import annotations
import argparse
import csv
import gc
import hashlib
import json
import platform
import time
import warnings
from pathlib import Path

import joblib
import numpy as np
import scipy
import sklearn
from sklearn.exceptions import ConvergenceWarning
from sklearn.metrics import f1_score
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.neural_network import MLPClassifier

from exp_a004_followup import Stage2Preprocessing
from exp_a004_official_mlp import load_inputs, checksum, CLASSES

CONDITIONS=("c9_only","raw_only","mix_50_50","raw_plus_c9","raw_twice")
EPOCHS={"c9_only":200,"raw_only":200,"mix_50_50":200,
        "raw_plus_c9":100,"raw_twice":100}
LABELS="0123456789ABCDEF"
PAIRS=(("B","8"),("8","B"),("D","0"),("0","D"),("5","F"),("F","5"))

def save_csv(path,rows):
    if not rows: raise ValueError("No rows for "+str(path))
    with Path(path).open("w",encoding="utf-8-sig",newline="") as f:
        writer=csv.DictWriter(f,fieldnames=list(rows[0]))
        writer.writeheader()
        writer.writerows(rows)

def group_rows(condition,seed,dataset,representation,truth,pred,writers=None):
    truth=np.asarray(truth,dtype=int)
    pred=np.asarray(pred,dtype=int)
    group={"all":np.ones(len(truth),dtype=bool),
           "digits":truth<10,"letters":truth>=10}
    for i,cl in enumerate(LABELS):
        group["class_"+cl]=(truth==i)
    if writers is not None:
        for w in [41,42,43,44]:
            group[f"writer_{w}"]=(writers==w)
            group[f"writer_{w}_digits"]=(writers==w)&(truth<10)
            group[f"writer_{w}_letters"]=(writers==w)&(truth>=10)
    out=[]
    for name,mask in group.items():
        if not mask.any():continue
        if name.startswith("class_"):
            ids=[LABELS.index(name[-1])]
        elif name=="digits" or name.endswith("_digits"):
            ids=list(range(10))
        elif name=="letters" or name.endswith("_letters"):
            ids=list(range(10,16))
        else:ids=list(range(16))
        t,p=truth[mask],pred[mask]
        out.append({"condition":condition,"seed":seed,"dataset":dataset,
                    "input":representation,"group":name,"n":len(t),
                    "correct":int(np.sum(t==p)),
                    "accuracy":float(np.mean(t==p)),
                    "macro_f1":float(f1_score(t,p,average="macro",
                                        labels=ids,zero_division=0))})
    return out

def train(condition,out,repo_root):
    if condition not in CONDITIONS:raise ValueError(condition)
    out=Path(out)
    if out.exists():raise FileExistsError(out)
    out.mkdir(parents=True)
    root=Path(repo_root).resolve()
    x_raw,y_train,x_val,y_val,x_hm,y_hm,writers,positions=load_inputs(root)
    assert x_raw.shape==(76800,784) and x_val.shape==(19200,784)
    assert x_hm.shape==(2560,784)
    c9=Stage2Preprocessing("C9_BBoxLongside20").fit(x_raw)
    c9_info=c9.description()
    print("EXP009 ORIGINAL C9 FIT: "+json.dumps(c9_info),flush=True)
    if c9_info["shape"]["median"]!=20.0:
        raise AssertionError("Original HypA C9 target differs from 20px")
    x_c9=c9.transform(x_raw)
    v_c9=c9.transform(x_val)
    h_c9=c9.transform(x_hm)

    # Keep labels/sample index unchanged. All normalized C9 pixels are copied
    # from the ORIGINAL HypA Stage2 implementation, with final CoM.
    raw=np.asarray(x_raw,dtype=np.float32)/255.0
    adjusted=x_c9/255.0
    if condition=="c9_only":
        train_x=adjusted
        train_y=y_train
    elif condition=="raw_only":
        train_x=raw
        train_y=y_train
    elif condition=="mix_50_50":
        splitter=StratifiedShuffleSplit(n_splits=1,test_size=.5,random_state=20261116)
        _,c9_indices=next(splitter.split(raw,y_train))
        mask=np.zeros(len(raw),dtype=bool)
        mask[c9_indices]=True
        train_x=raw.copy()
        train_x[mask]=adjusted[mask]
        train_y=y_train
        assert mask.sum()==38400
        del mask,c9_indices
    elif condition=="raw_plus_c9":
        train_x=np.concatenate([raw,adjusted],axis=0)
        train_y=np.concatenate([y_train,y_train],axis=0)
    elif condition=="raw_twice":
        train_x=np.concatenate([raw,raw],axis=0)
        train_y=np.concatenate([y_train,y_train],axis=0)
    assert len(train_x)==(153600 if condition in ("raw_plus_c9","raw_twice") else 76800)
    assert len(train_y)==len(train_x)
    assert np.isfinite(train_x).all() and 0<=train_x.min()<=train_x.max()<=1.0

    evaluation={
        ("OfficialVal","raw"):(x_val/255.0,y_val,None,None),
        ("OfficialVal","c9"):(v_c9/255.0,y_val,None,None),
        ("Handmade","raw"):(x_hm/255.0,y_hm,writers,positions),
        ("Handmade","c9"):(h_c9/255.0,y_hm,writers,positions),
    }
    metrics=[]
    predictions=[]
    pair_errors=[]
    fits=[]
    source_path=root/"02_data/datasets/MNIST_EMNIST_official_split_112000.npz"
    for seed in [42,43,44]:
        model=MLPClassifier(hidden_layer_sizes=(100,),activation="relu",
            solver="adam",alpha=0.0001,learning_rate_init=0.001,
            max_iter=EPOCHS[condition],tol=0.0,
            n_iter_no_change=201,batch_size="auto",shuffle=True,
            random_state=seed)
        tic=time.monotonic()
        print(f"START EXP009 condition={condition} seed={seed} "
              f"samples={len(train_y)} max_iter={model.max_iter}",flush=True)
        with warnings.catch_warnings():
            warnings.simplefilter("ignore",ConvergenceWarning)
            model.fit(train_x,train_y)
        duration=time.monotonic()-tic
        for (dataset,representation),(X,truth,ws,ids) in evaluation.items():
            pred=model.predict(X).astype(int)
            metrics.extend(group_rows(condition,seed,dataset,representation,
                                      truth,pred,ws))
            if dataset=="Handmade":
                for i in range(len(truth)):
                    predictions.append({"condition":condition,"seed":seed,
                        "input":representation,"writer":int(ws[i]),
                        "index_in_writer":int(ids[i]),
                        "actual":LABELS[int(truth[i])],
                        "predicted":LABELS[int(pred[i])]})
                for t,p in PAIRS:
                    pair_errors.append({"condition":condition,"seed":seed,
                      "input":representation,"actual":t,"predicted":p,
                      "count":int(np.sum((truth==LABELS.index(t))&
                                              (pred==LABELS.index(p))))})
            overall=np.mean(truth==pred)
            print(f"RESULT {condition} seed={seed} {dataset}/{representation}: "
                  f"{overall*100:.4f}%",flush=True)
        model_file=out/f"mlp_{condition}_seed{seed}.joblib"
        joblib.dump(model,model_file,compress=2)
        fits.append({"condition":condition,"seed":seed,
                     "train_examples":len(train_y),"epochs_limit":EPOCHS[condition],
                     "minibatch_updates_nominal":EPOCHS[condition]*
                       ((len(train_y)+199)//200),
                     "actual_epochs":int(model.n_iter_),
                     "train_loss":float(model.loss_),
                     "train_seconds":round(duration,1),
                     "model_sha256":checksum(model_file)})
        save_csv(out/"group_metrics.csv",metrics)
        save_csv(out/"handmade_predictions.csv",predictions)
        save_csv(out/"pair_errors.csv",pair_errors)
        save_csv(out/"fit_metrics.csv",fits)
        print(f"FIT END {condition} seed={seed} "
              f"epoch={model.n_iter_} time={duration:.1f}s",flush=True)
        del model
        gc.collect()
    manifest={"experiment":"Normal-HypB-EXP009",
        "condition":condition,"seeds":[42,43,44],
        "study":"Original HypA C9; raw+C9 Data Augmentation",
        "training_n":len(train_y),"original_unique_n":76800,
        "epochs_limit":EPOCHS[condition],"batch_size_auto":200,
        "nominal_gradient_updates":EPOCHS[condition]*((len(train_y)+199)//200),
        "fixed_aug_selection_seed":20261116,
        "original_hypa_c9":c9_info,
        "source_sha256":checksum(source_path),
        "python_version":platform.python_version(),
        "numpy_version":np.__version__,"scipy_version":scipy.__version__,
        "sklearn_version":sklearn.__version__,
        "original_test_used":False,"professor_test_used":False,
        "handmade_used_for_fitting":False,
        "evaluation_inputs":["raw","c9"],"evaluation_datasets":["OfficialVal","Handmade"],
        "fit_metrics":fits}
    (out/"run_manifest.json").write_text(json.dumps(manifest,indent=2,ensure_ascii=False),
                                         encoding="utf-8")
    print("COMPLETE EXP009 "+condition+" all 3 models trained",flush=True)

def read_csv(p):
    with Path(p).open("r",encoding="utf-8-sig",newline="") as f:
        return list(csv.DictReader(f))

def aggregate(root,out):
    root=Path(root)
    out=Path(out)
    if out.exists():raise FileExistsError(out)
    out.mkdir(parents=True)
    files={}
    grouped=[]
    preds={}
    manifests={}
    for condition in CONDITIONS:
        folder=root/("exp009-"+condition)
        manifest=json.loads((folder/"run_manifest.json").read_text(encoding="utf-8"))
        assert manifest["condition"]==condition
        assert not manifest["original_test_used"] and not manifest["professor_test_used"]
        assert not manifest["handmade_used_for_fitting"]
        assert manifest["nominal_gradient_updates"]==76800
        manifests[condition]=manifest
        rows=read_csv(folder/"group_metrics.csv")
        for s in [42,43,44]:
            for dataset in ["OfficialVal","Handmade"]:
                for input_kind in ["raw","c9"]:
                    for group in ["all","digits","letters"]:
                        if len([r for r in rows if r["seed"]==str(s) and
                            r["dataset"]==dataset and r["input"]==input_kind and
                            r["group"]==group])!=1:
                            raise AssertionError(f"Missing metric: {condition} {s} {dataset} {input_kind} {group}")
        grouped.extend(rows)
        pred=read_csv(folder/"handmade_predictions.csv")
        for seed in [42,43,44]:
            for inp in ["raw","c9"]:
                subset=[r for r in pred if r["seed"]==str(seed) and r["input"]==inp]
                assert len(subset)==2560
                key_set={(r["writer"],r["index_in_writer"]) for r in subset}
                assert len(key_set)==2560
                preds[condition,seed,inp]={ (r["writer"],r["index_in_writer"]):r
                                           for r in subset }
        files[condition]=folder

    # Preserve raw metrics, produce group summary and paired per-image changes.
    save_csv(out/"all_group_metrics.csv",grouped)
    summaries=[]
    for cond in CONDITIONS:
        for dataset in ["OfficialVal","Handmade"]:
            for inp in ["raw","c9"]:
                for group in ["all","digits","letters"]:
                    xs=[r for r in grouped if r["condition"]==cond and
                        r["dataset"]==dataset and r["input"]==inp and r["group"]==group]
                    if len(xs)!=3:raise AssertionError("Missing 3-seed row")
                    summaries.append({"condition":cond,"dataset":dataset,
                      "input":inp,"group":group,"n_per_seed":xs[0]["n"],
                      "seed_count":3,
                      "accuracy_mean":float(np.mean([float(r["accuracy"]) for r in xs])),
                      "accuracy_min":min(float(r["accuracy"]) for r in xs),
                      "accuracy_max":max(float(r["accuracy"]) for r in xs),
                      "macro_f1_mean":float(np.mean([float(r["macro_f1"]) for r in xs]))})
    save_csv(out/"three_seed_summary.csv",summaries)
    paired=[]
    for condition in CONDITIONS:
        if condition=="c9_only":continue
        for seed in [42,43,44]:
            for inp in ["raw","c9"]:
                source=preds["c9_only",seed,inp]
                candidate=preds[condition,seed,inp]
                for group in ["all","digits","letters"]:
                    keys=[k for k in source if group=="all" or
                          ((source[k]["actual"] in "0123456789")==(group=="digits"))]
                    fixed=broken=0
                    for key in keys:
                        a,b=source[key],candidate[key]
                        if a["actual"]!=b["actual"]:
                            raise AssertionError("Truth label mismatch")
                        ac=a["actual"]==a["predicted"]
                        bc=b["actual"]==b["predicted"]
                        fixed+=int(not ac and bc)
                        broken+=int(ac and not bc)
                    paired.append({"condition":condition,"seed":seed,
                      "input":inp,"group":group,"n":len(keys),
                      "corrected_from_c9":fixed,"new_wrong_vs_c9":broken,
                      "net_correct":fixed-broken,
                      "delta_acc_pp":100*(fixed-broken)/len(keys)})
    save_csv(out/"paired_handmade_vs_c9.csv",paired)
    def pct(c,d,k,g):
        rows=[r for r in summaries if r["condition"]==c and r["dataset"]==d
              and r["input"]==k and r["group"]==g]
        assert len(rows)==1
        return f"{100*rows[0]['accuracy_mean']:.4f}%"
    result=["# Normal 가설 B EXP유009 — 원본 + C9 Data Augmentation 실측",
        "",
        "**상태: 공식 Train 재학습 및 3-Seed 평가 완료.**",
        "",
        "- HypA EXP004 original C9 implementation and same train split.",
        "- All 5 methods fit only on official Train, no original/professor Test, no handmade training.",
        "- Training minibatch update budget **matched at 76,800 nominal steps** (76800×200 epochs vs 153600×100 epochs).",
        "- Train original vs C9 selection 50:50 uses fixed StratifiedShuffleSplit seed=20261116.",
        "- Both raw and C9 inference separately evaluated, no post-hoc selection on handmade.",
        "",
        "## Evaluation on C9-normalized inputs",
        "",
        "| Training | Official Val | Handmade all | Digits | Letters |",
        "|---|---:|---:|---:|---:|"]
    for c in CONDITIONS:
        result.append(f"| {c} | {pct(c,'OfficialVal','c9','all')} | "
          f"{pct(c,'Handmade','c9','all')} | {pct(c,'Handmade','c9','digits')} | "
          f"{pct(c,'Handmade','c9','letters')} |")
    result+=["","## Evaluation on original raw inputs","","| Training | Official Val | Handmade all | Digits | Letters |",
              "|---|---:|---:|---:|---:|"]
    for c in CONDITIONS:
        result.append(f"| {c} | {pct(c,'OfficialVal','raw','all')} | "
          f"{pct(c,'Handmade','raw','all')} | {pct(c,'Handmade','raw','digits')} | "
          f"{pct(c,'Handmade','raw','letters')} |")
    result+=["","## Interpretation",
      "- Differences are based on actual 3-seed model training in the same workflow.",
      "- C9-only versus raw+C9 and raw twice isolates added transform diversity from mere sample repeats under matched nominal step budget.",
      "- Each 153600 set contains **the same 76800 underlying official training images** twice, with an alternate view; it is not independent new data.",
      "- The four writers in handmade have been used in development repeatedly; do not treat their scores as independent test performance.",
      "- Official Validation threshold ≥97.30% and handmade digits ≥85%, letters ≥80% are simultaneous criteria.",
      "- No final Pipeline change and no main merge are implied."
    ]
    (out/"SUMMARY.md").write_text("\n".join(result)+"\n",encoding="utf-8")
    print("EXP009 SUMMARY COMPLETED",flush=True)
    print("\n".join(result),flush=True)

def main():
    p=argparse.ArgumentParser()
    p.add_argument("--repo-root",default=".")
    p.add_argument("--condition",choices=CONDITIONS)
    p.add_argument("--output",type=Path,required=True)
    p.add_argument("--aggregate",action="store_true")
    p.add_argument("--input-root",type=Path)
    args=p.parse_args()
    if args.aggregate:
        if args.input_root is None:p.error("--aggregate requires input-root")
        aggregate(args.input_root,args.output)
    else:
        if args.condition is None:p.error("Need --condition")
        train(args.condition,args.output,args.repo_root)

if __name__=="__main__":
    main()
```

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp_a004_followup.py
#!/usr/bin/env python3
"""EXP유004 continued experiment: train-fitted morphology and size variants.

Same official Train, same 3 seeds, same MLP fit and pass criteria as Stage 1.
No writer/label information in preprocessing. No official or professor Test.
"""
from __future__ import annotations
import argparse
from pathlib import Path
import numpy as np
import cv2
from stroke_normalization import StrokeWidthNormalizer, grayscale, topology, center_of_mass
from width_normalization import bounds
import exp_a004_official_mlp as base

CONDITIONS = ("C5_Stroke045", "C6_Stroke060", "C7_MedianWidth035",
              "C8_Stroke030_MedianWidth035", "C9_BBoxLongside20")

def bbox_warp(img, scale_x, scale_y):
    mask = bounds(img, 64)
    if mask is None:
        return img.copy()
    cx, cy = (mask[0] + mask[1])/2, (mask[2] + mask[3])/2
    matrix = np.array([[scale_x, 0, (1-scale_x)*cx],
                       [0, scale_y, (1-scale_y)*cy]], np.float32)
    return cv2.warpAffine(img, matrix, (28,28), flags=cv2.INTER_LINEAR,
                          borderMode=cv2.BORDER_CONSTANT, borderValue=0)

class SizeToTrainingMedian:
    def __init__(self, kind, strength=.35, threshold=64, preserve_topology=True):
        self.kind = kind
        self.strength = strength
        self.threshold = threshold
        self.preserve_topology = preserve_topology

    def _measured(self, im):
        b = bounds(im, self.threshold)
        if b is None:
            return None
        w, h = b[1] - b[0] + 1, b[3] - b[2] + 1
        return w if self.kind == "width" else max(w, h)

    def fit(self, x):
        vals = [self._measured(im) for im in grayscale(x)]
        vals = np.array([v for v in vals if v is not None], float)
        if len(vals) < 20:
            raise RuntimeError("Insufficient official training images")
        self.median_ = float(np.median(vals))
        self.q20_, self.q80_ = map(float, np.quantile(vals, [.2, .8]))
        self.fit_count_ = int(len(vals))
        return self

    def transform(self, x):
        if not hasattr(self, "median_"):
            raise RuntimeError("Fit reference on Training first")
        a = grayscale(x)
        out = np.empty_like(a)
        for i, im in enumerate(a):
            current = self._measured(im)
            updated = im.copy()
            if current and self.strength > 0:
                desired = 1.0 + self.strength * (self.median_ / current - 1)
                desired = float(np.clip(desired, .75, 1.25))
                if abs(desired - 1.0) > 1e-4:
                    before = topology(im)
                    for _ in range(5):
                        sx, sy = (desired, 1.0) if self.kind == "width" else (desired, desired)
                        candidate = bbox_warp(im, sx, sy)
                        if not self.preserve_topology or topology(candidate) == before:
                            updated = candidate
                            break
                        desired = 1 + (desired-1)/2
            out[i] = updated
        return out.reshape(len(a),784)

class Stage2Preprocessing:
    def __init__(self, policy):
        if policy not in CONDITIONS:
            raise ValueError(policy)
        self.policy = policy
        self.stroke = None
        self.shape = None

    def fit(self, X):
        self.n_fit = len(X)
        if self.policy in ("C5_Stroke045", "C6_Stroke060", "C8_Stroke030_MedianWidth035"):
            strength = .45 if self.policy == "C5_Stroke045" else .60 if self.policy == "C6_Stroke060" else .30
            self.stroke = StrokeWidthNormalizer(strength=strength,
                                                 preserve_topology=True,
                                                 apply_center=False).fit(X)
        if self.policy in ("C7_MedianWidth035", "C8_Stroke030_MedianWidth035", "C9_BBoxLongside20"):
            kind = "longside" if self.policy == "C9_BBoxLongside20" else "width"
            strength = 1.0 if kind == "longside" else .35
            train = self.stroke.transform(X) if self.stroke else X
            self.shape = SizeToTrainingMedian(kind=kind, strength=strength,
                                              preserve_topology=True).fit(train)
        return self

    def transform(self, X):
        x = grayscale(X)
        if self.stroke:
            x = self.stroke.transform(x).reshape(-1,28,28)
        if self.shape:
            x = self.shape.transform(x).reshape(-1,28,28)
        result = np.empty((len(x),784),np.float32)
        for i,im in enumerate(x):
            result[i] = center_of_mass(im).reshape(-1)
        return result

    def description(self):
        info = {"condition":self.policy,"training_fit_count":self.n_fit,
                "fit_data":"OfficialTrainOnly","uses_labels":False,
                "uses_writer":False,"com_last":True,"topology_guard":True}
        if self.stroke:
            t=self.stroke
            info["stroke"]={"strength":t.strength,"median":float(t.median_),
                            "q20":float(t.q_low_),"q80":float(t.q_high_)}
        if self.shape:
            t=self.shape
            info["shape"]={"kind":t.kind,"strength":t.strength,
                           "median":t.median_,"q20":t.q20_,"q80":t.q80_}
        return info

def test_preprocessors():
    x=np.zeros((40,28,28),np.float32)
    for i in range(40):
        cv2.line(x[i],(4+i%8,4),(4+i%8,22),255,1+i%5)
    for condition in CONDITIONS:
        p=Stage2Preprocessing(condition).fit(x)
        t=p.transform(x)
        assert t.shape==(40,784) and np.isfinite(t).all()
        assert 0 <= t.min() <= t.max() <= 255
        assert np.array_equal(t,p.transform(x))
    print("STAGE2 SYNTHETIC SELF TEST PASSED; no real training claimed",flush=True)

if __name__=="__main__":
    parser=argparse.ArgumentParser()
    parser.add_argument("--repo-root",type=Path)
    parser.add_argument("--out-dir",type=Path)
    parser.add_argument("--condition",choices=CONDITIONS)
    parser.add_argument("--seeds",nargs="+",type=int,default=[42,43,44])
    parser.add_argument("--max-iter",type=int,default=200)
    parser.add_argument("--self-test",action="store_true")
    args=parser.parse_args()
    if args.self_test:
        test_preprocessors()
    else:
        if args.repo_root is None or args.out_dir is None or args.condition is None:
            parser.error("repo-root, out-dir, condition required")
        # The original runner performs the identical official SHA/split check and
        # identical MLP training, metrics, model export, and label-blind validation.
        base.Preprocessing = Stage2Preprocessing
        base.run(args.repo_root,args.out_dir,args.condition,args.seeds,args.max_iter)
```

## stroke_normalization

원본: [05_model_optimize/MLP/Normal/scripts/stroke_normalization.py](https://github.com/ryu980920/Machine_Learning/blob/b5810b1e677fdd4588b0c20df480aa83168df938/05_model_optimize/MLP/Normal/scripts/stroke_normalization.py)

Training-fitted, class-blind 28x28 grayscale stroke width normalization.



```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/stroke_normalization.py
#!/usr/bin/env python3
"""Training-fitted, class-blind 28x28 grayscale stroke width normalization."""
from __future__ import annotations
import cv2
import numpy as np
from scipy.ndimage import distance_transform_edt, label
from skimage.morphology import skeletonize
from skimage.measure import euler_number
from sklearn.base import BaseEstimator, TransformerMixin

KERNEL = cv2.getStructuringElement(cv2.MORPH_CROSS, (3, 3))

def grayscale(images):
    a = np.asarray(images)
    if a.ndim == 2 and a.shape[1] == 784:
        a = a.reshape(-1, 28, 28)
    elif a.ndim == 2 and a.shape == (28, 28):
        a = a[None]
    if a.ndim != 3 or a.shape[1:] != (28, 28):
        raise ValueError("Requires 28x28 grayscale images")
    if not np.isfinite(a).all() or a.min() < 0 or a.max() > 255:
        raise ValueError("Pixel values outside 0..255")
    return a.astype(np.float32, copy=False)

def stroke_width(img, threshold=64):
    mask = np.asarray(img) >= threshold
    if not mask.any():
        return np.nan
    radii = distance_transform_edt(mask)[skeletonize(mask)]
    return max(1., float(2 * np.median(radii) - 1)) if len(radii) else np.nan

def topology(img, threshold=128):
    mask = np.asarray(img) >= threshold
    return (int(label(mask, structure=np.ones((3, 3)))[1]),
            int(euler_number(mask, connectivity=2)))

def center_of_mass(img, threshold=32):
    im = np.asarray(img, dtype=np.float32)
    weighted = np.where(im >= threshold, im, 0)
    mass = float(weighted.sum())
    if mass == 0:
        return im.copy()
    yy, xx = np.indices((28, 28))
    dx = int(np.rint(13.5 - float((weighted * xx).sum()) / mass))
    dy = int(np.rint(13.5 - float((weighted * yy).sum()) / mass))
    out = np.zeros_like(im)
    sy0, sy1 = max(0, -dy), min(28, 28 - dy)
    sx0, sx1 = max(0, -dx), min(28, 28 - dx)
    if sy0 < sy1 and sx0 < sx1:
        out[sy0 + dy:sy1 + dy, sx0 + dx:sx1 + dx] = im[sy0:sy1, sx0:sx1]
    return out

class StrokeWidthNormalizer(BaseEstimator, TransformerMixin):
    def __init__(self, strength=.15, threshold=64, fit_quantiles=(.2, .8),
                 preserve_topology=True, apply_center=False):
        self.strength = strength
        self.threshold = threshold
        self.fit_quantiles = fit_quantiles
        self.preserve_topology = preserve_topology
        self.apply_center = apply_center

    def fit(self, X, y=None):
        if not 0 <= self.strength <= 1:
            raise ValueError("Invalid stroke strength")
        vals = np.array([stroke_width(im, self.threshold) for im in grayscale(X)], dtype=float)
        vals = vals[np.isfinite(vals)]
        if len(vals) < 20:
            raise ValueError("Not enough training strokes")
        self.q_low_, self.q_high_ = np.quantile(vals, self.fit_quantiles)
        self.median_ = float(np.median(vals))
        self.n_fitted_ = len(vals)
        return self

    def transform(self, X):
        if not hasattr(self, "q_low_"):
            raise RuntimeError("Fit on official Training only")
        src = grayscale(X)
        out = np.empty_like(src)
        for i, img in enumerate(src):
            width = stroke_width(img, self.threshold)
            updated = img.copy()
            if np.isfinite(width) and self.strength > 0:
                target = (cv2.dilate(img, KERNEL) if width < self.q_low_ else
                          cv2.erode(img, KERNEL) if width > self.q_high_ else img)
                if target is not img:
                    orig_topo = topology(img)
                    alpha = self.strength
                    for _ in range(5):
                        test = np.clip(img * (1 - alpha) + target * alpha, 0, 255)
                        if not self.preserve_topology or topology(test) == orig_topo:
                            updated = test
                            break
                        alpha /= 2
            out[i] = center_of_mass(updated) if self.apply_center else updated
        return out.reshape(len(src), 784)
```

원본: [05_model_optimize/MLP/Normal/scripts/width_normalization.py](https://github.com/ryu980920/Machine_Learning/blob/b5810b1e677fdd4588b0c20df480aa83168df938/05_model_optimize/MLP/Normal/scripts/width_normalization.py)  
Git blob: `c5f1757b2fb23b6fd3ca48df422719f0b1fcc3a0`. [기록 및 결과](https://github.com/ryu980920/Machine_Learning/tree/b5810b1e677fdd4588b0c20df480aa83168df938/05_model_optimize/MLP/Normal/logs)

Training-fitted, class-blind glyph horizontal extent normalization.



```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/width_normalization.py
#!/usr/bin/env python3
"""Training-fitted, class-blind glyph horizontal extent normalization."""
from __future__ import annotations
import cv2
import numpy as np
from sklearn.base import BaseEstimator, TransformerMixin
from stroke_normalization import grayscale, center_of_mass, topology

def bounds(img, threshold=64):
    yy, xx = np.where(np.asarray(img) >= threshold)
    if len(xx) == 0:
        return None
    return int(xx.min()), int(xx.max()), int(yy.min()), int(yy.max())

class GlyphWidthNormalizer(BaseEstimator, TransformerMixin):
    def __init__(self, strength=.15, threshold=64, quantiles=(.2, .8),
                 preserve_topology=True, apply_center=False):
        self.strength = strength
        self.threshold = threshold
        self.quantiles = quantiles
        self.preserve_topology = preserve_topology
        self.apply_center = apply_center

    def fit(self, X, y=None):
        if not 0 <= self.strength <= 1:
            raise ValueError("Invalid width strength")
        widths = []
        for im in grayscale(X):
            b = bounds(im, self.threshold)
            if b is not None:
                widths.append(b[1] - b[0] + 1)
        if len(widths) < 20:
            raise ValueError("Not enough training glyphs")
        self.q_low_, self.q_high_ = np.quantile(widths, self.quantiles)
        self.median_ = float(np.median(widths))
        self.n_fitted_ = len(widths)
        return self

    def transform(self, X):
        if not hasattr(self, "q_low_"):
            raise RuntimeError("Fit on official Training only")
        src = grayscale(X)
        out = np.empty_like(src)
        for i, img in enumerate(src):
            b = bounds(img, self.threshold)
            result = img.copy()
            if b is not None and self.strength > 0:
                w = b[1] - b[0] + 1
                goal = self.q_low_ if w < self.q_low_ else self.q_high_ if w > self.q_high_ else w
                if goal != w:
                    scale = float(np.clip(1 + self.strength * (goal / w - 1), .85, 1.15))
                    cx = (b[0] + b[1]) / 2
                    before = topology(img)
                    for _ in range(5):
                        matrix = np.array([[scale, 0, (1 - scale) * cx], [0, 1, 0]], np.float32)
                        changed = cv2.warpAffine(img, matrix, (28, 28),
                                                 flags=cv2.INTER_LINEAR,
                                                 borderMode=cv2.BORDER_CONSTANT, borderValue=0)
                        if not self.preserve_topology or topology(changed) == before:
                            result = changed
                            break
                        scale = 1 + (scale - 1) / 2
            out[i] = center_of_mass(result) if self.apply_center else result
        return out.reshape(len(src), 784)
```

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp_a004_aggregate.py
#!/usr/bin/env python3
"""Aggregate EXP유004 real 15 fitted MLP outputs. Do not select a Test-tuned policy."""
from __future__ import annotations
import csv
import json
from pathlib import Path
from statistics import mean, pstdev

CONDITIONS = ["C0_CoM", "C1_Stroke015", "C2_Stroke030",
              "C3_GlyphWidth015", "C4_Stroke015_GlyphWidth015"]
SEEDS = [42, 43, 44]
TARGETS = {"Handmade/letters": .80, "Handmade/digits": .85,
           "OfficialVal/all": .973}

def read_csv(path):
    with open(path, encoding="utf-8-sig", newline="") as f:
        return list(csv.DictReader(f))

def save_csv(path, rows):
    with open(path, "w", encoding="utf-8-sig", newline="") as f:
        w = csv.DictWriter(f, fieldnames=list(rows[0]))
        w.writeheader()
        w.writerows(rows)

def main(root):
    records = []
    all_pairs, all_groups = [], []
    for condition in CONDITIONS:
        folder = root / ("normal-a004-" + condition)
        manifest = json.loads((folder / "run_manifest.json").read_text(encoding="utf-8"))
        if manifest["condition"] != condition or manifest["official_original_test_used"] or manifest["professor_test_used"]:
            raise RuntimeError("Manifest failed for " + condition)
        if sorted(int(r["seed"]) for r in manifest["seed_results"]) != SEEDS:
            raise RuntimeError("Missing fitted seed(s) in " + condition)
        groups = read_csv(folder / "group_metrics.csv")
        pairs = read_csv(folder / "pair_errors.csv")
        if len(pairs) != len(SEEDS) * 6:
            raise RuntimeError("Missing confusion pairs")
        all_groups.extend(groups)
        all_pairs.extend(pairs)
        for seed in SEEDS:
            def find(dataset, group):
                matches = [r for r in groups if int(r["seed"]) == seed and
                           r["dataset"] == dataset and r["group"] == group]
                if len(matches) != 1:
                    raise RuntimeError(f"Ambiguous group {condition} {seed} {dataset}/{group}")
                return float(matches[0]["accuracy"])
            letters = find("Handmade", "letters")
            digits = find("Handmade", "digits")
            official = find("OfficialVal", "all")
            error_sum = sum(int(p["count"]) for p in pairs if int(p["seed"]) == seed)
            records.append({"condition": condition, "seed": seed, "letters": letters,
                            "digits": digits, "official_val": official,
                            "six_directional_errors": error_sum,
                            "pass_all_three": (letters >= .80 and digits >= .85 and official >= .973)})
    save_csv(root / "all_group_metrics.csv", all_groups)
    save_csv(root / "all_pair_errors.csv", all_pairs)
    save_csv(root / "pass_by_seed.csv", records)
    averages = []
    for condition in CONDITIONS:
        rows = [r for r in records if r["condition"] == condition]
        m = {k: mean(r[k] for r in rows) for k in ("letters", "digits", "official_val", "six_directional_errors")}
        s = {k + "_sd": pstdev(r[k] for r in rows) for k in ("letters", "digits", "official_val")}
        passed = (m["letters"] >= .80 and m["digits"] >= .85 and m["official_val"] >= .973)
        consistent = sum(bool(r["pass_all_three"]) for r in rows) >= 2
        averages.append({"condition": condition, **m, **s,
                         "seeds_individual_pass": sum(bool(r["pass_all_three"]) for r in rows),
                         "development_pass": bool(passed and consistent)})
    save_csv(root / "method_summary.csv", averages)
    lines = [
        "# 가설 A · EXP유004 — 공식 Training 재학습 실측 (검토용)",
        "",
        "**모델:** MLPClassifier 100 ReLU hidden nodes, Adam, 공식 Training 76,800장만 학습.",
        "**전처리:** 공식 Training에서만 굵기/가로폭 기준을 fit. 클래스·작성자·정답 사용 없음.",
        "**반복:** 조건별 seed 42, 43, 44. 5조건 × 3회 = 15개 MLP.",
        "**Test:** 공식 Original Test와 교수님 Test는 미사용. Hand-made는 개발용 평가에만 사용.",
        "",
        "통과 판정은 **3개 seed 평균이 문자≥80%, 숫자≥85%, 공식 Validation≥97.30%**이고, **최소 2개 seed가 세 지표를 동시에 충족**해야 한다.",
        "이는 개발 통과이며, 미관찰 작성자 또는 최종 Test 성능을 의미하지 않는다.",
        "",
        "| 조건 | 문자 평균 | 숫자 평균 | 공식 Val 평균 | 6방향 오류 평균 | 개별 통과 seed | 개발 통과 |",
        "|---|---:|---:|---:|---:|---:|---|",
    ]
    for r in averages:
        lines.append(f'| {r["condition"]} | {r["letters"]*100:.3f}% | '
                     f'{r["digits"]*100:.3f}% | {r["official_val"]*100:.3f}% | '
                     f'{r["six_directional_errors"]:.1f} | '
                     f'{r["seeds_individual_pass"]}/3 | '
                     f'{"PASS" if r["development_pass"] else "FAIL"} |')
    lines += ["", "### 해석 범위",
              "- 수치가 높은 후보를 단일 seed로 사후 선택하지 않는다.",
              "- 각 방법의 실제 원본/변형 손글씨 PNG, 개별 seed 체크포인트 및 예측은 Actions Artifact에 보존한다.",
              "- 모델과 학습/추론 전처리 버전은 함께 보관해야 동일 결과를 재현할 수 있다.",
              "- 기존 Hand-made 4명에 대한 반복 분석이므로 최종 독립 Test의 통과는 별도로 필요하다."]
    (root / "SUMMARY.md").write_text("\n".join(lines) + "\n", encoding="utf-8")
    (root / "run_index.json").write_text(json.dumps({
        "conditions": CONDITIONS, "seeds": SEEDS, "trained_models": len(records),
        "official_original_test_used": False, "professor_test_used": False,
        "gate": TARGETS, "results": averages
    }, indent=2, ensure_ascii=False), encoding="utf-8")
    print("\n".join(lines), flush=True)

if __name__ == "__main__":
    import argparse
    p = argparse.ArgumentParser()
    p.add_argument("--root", type=Path, required=True)
    main(p.parse_args().root)
```
