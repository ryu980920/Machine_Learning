## 사용법
원본 코드 셀은 `%%archive_source`로 등록하는 실제 원본 코드이며, 함수명 충돌을 막기 위해 현재 커널에서 임의 실행하지 않는다. 맨 아래의 실행 셀에서 원본 스크립트/Notebook을 고정 커밋의 **별도 checkout**에서 실행한다. 각 실험의 실행 인자·입력·사전 조건은 위 원본 로그/코드로 확인한다. 기존 로그·결과와 새 출력이 섞이지 않도록 독립 복사본을 사용한다. Original Test/교수님 Test를 포함한 파일은 실행 환경에 두지 않는다. 과거 스크립트가 해당 자료를 요구할 경우 실행하지 않는다. 자동 재실행 및 성능 동등성은 미검증.

원본 실행을 위한 로그 디렉터리를 별도로 지정한다.

```python
# 공통 실행 제어 — Run All은 원본 코드 등록만 하며 학습하지 않는다.
from pathlib import Path
import hashlib, json, subprocess, sys, datetime
from IPython import get_ipython
SOURCE_CELLS = {}
def archive_source(line, cell):
    SOURCE_CELLS[line.strip()] = cell
get_ipython().register_magic_function(archive_source, magic_kind="cell", magic_name="archive_source")
# 실행 시에만 설정: 기존 작업 트리가 아닌 임시 checkout (git worktree add --detach ... <ref>)
WORKTREE = None
LOG_DIR = None
RUN_PLAN = []  # 예: [("원본/스크립트.py", ["--인자", "값"])]
ALLOW_EXECUTION = False

def blob_sha(data):
    return hashlib.sha1(f"blob {len(data)}\0".encode()+data).hexdigest()

def get_verified_source(path):
    if WORKTREE is None:
        raise ValueError("해당 커밋의 별도 checkout 경로 WORKTREE를 지정하세요.")
    root=Path(WORKTREE).expanduser().resolve()
    entry=SOURCE_INDEX[path]
    head=subprocess.check_output(["git","-C",str(root),"rev-parse","HEAD"],text=True).strip()
    if head != entry["ref"]:
        raise ValueError(f"checkout ref 불일치: {head} != {entry['ref']}")
    p=root/path
    if blob_sha(p.read_bytes()) != entry["sha"]:
        raise ValueError(f"원본 Git blob 불일치: {path}")
    if path in SOURCE_CELLS:
        if p.read_text(encoding="utf-8-sig").rstrip("\n") != SOURCE_CELLS[path].rstrip("\n").lstrip("\ufeff"):
            raise ValueError(f"Notebook 보존 코드와 원본이 다릅니다: {path}")
    return root,p

def show_original(path):
    from IPython.display import display, Code
    _,p=get_verified_source(path)
    display(Code(p.read_text(encoding="utf-8"),language="python" if p.suffix==".py" else "json"))

def execute_original(path, args):
    if not ALLOW_EXECUTION:
        raise RuntimeError("명시적 실행 승인이 필요합니다.")
    root,p=get_verified_source(path)
    if LOG_DIR is None:
        raise ValueError("새 외부 LOG_DIR를 지정하세요.")
    logs=Path(LOG_DIR).expanduser().resolve();logs.mkdir(parents=True,exist_ok=True)
    if logs==root or root in logs.parents:
        raise ValueError("LOG_DIR는 checkout 밖에 두세요.")
    stamp=datetime.datetime.now(datetime.timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
    if p.suffix==".ipynb":
        cmd=[sys.executable,"-m","jupyter","nbconvert","--to","notebook","--execute",str(p),"--output-dir",str(logs),"--output",f"{p.stem}_{stamp}.ipynb",*args]
    else:
        cmd=[sys.executable,str(p),*args]
    log=logs/f"{p.stem}_{stamp}.log"
    with log.open("x",encoding="utf-8") as f:
        f.write(json.dumps({"source":path,**SOURCE_INDEX[path],"command":cmd,"cwd":str(p.parent)},ensure_ascii=False)+"\n");f.flush()
        result=subprocess.run(cmd,cwd=p.parent,stdout=f,stderr=subprocess.STDOUT)
        f.write(f"\nreturncode={result.returncode}\n")
    print(log)
    result.check_returncode()
```

EXP014: class-family-independent size augmentation and stroke controls.
Three new MLPC classifiers per seed, each fits 76,800 official Train source images.
Archived EXP006 CoM/SizeCoM models serve as frozen controls; never use test sets.

CLI 인자 (정적 추출): `--seed`, `--models-root`, `--out`

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp014_balanced_augmentation.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""EXP014: class-family-independent size augmentation and stroke controls.
Three new MLPC classifiers per seed, each fits 76,800 official Train source images.
Archived EXP006 CoM/SizeCoM models serve as frozen controls; never use test sets.
"""
from __future__ import annotations
import argparse,csv,gzip,json,time,warnings
from pathlib import Path
import joblib, cv2
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix
from exp006_size_com_comparison import (
    ROOT,WRITERS,LABELS,PixelScaler255,load_official,load_writers,
    image_geometry,apply_size,process,save_csv,sha256)
from exp011_012_causal_probes import fixed_scale
from stroke_normalization import topology,KERNEL

VARIANTS=("balanced_scale","balanced_scale_stroke","sizecom_balanced_scale")
SCALES=(.8,.9,1.,1.1,1.2)
PAIRS=(("B","8"),("8","B"),("D","0"),("0","D"),("F","5"),("5","F"))

def balanced_indices(y,seed=20261116):
    rng=np.random.default_rng(seed)
    selections=[]; targets=np.zeros(len(y),np.int32)
    for cls in range(16):
        orig=np.flatnonzero(y==cls)
        # Each official source class can have an uneven count in this split.
        # Sample near one-half, rounded DOWN to a multiple of five, so EACH
        # target size is assigned exactly equally within EACH true class.
        n_aug=(len(orig)//10)*5
        if n_aug<5: raise AssertionError("insufficient per-class training data")
        sel=np.sort(rng.choice(orig,size=n_aug,replace=False))
        choices=np.tile(np.array([16,18,20,22,24],dtype=np.int32),
                        len(sel)//5)
        rng.shuffle(choices)
        targets[sel]=choices
        selections.extend(sel.tolist())
    ids=np.array(sorted(selections),dtype=np.int32)
    assert .48 <= len(ids)/len(y) <= .50
    assert set(np.unique(targets[ids]))=={16,18,20,22,24}
    return ids,targets

def stroke_jitter(images,ids):
    """Conservative, topology-checked gray morphological jitter (train only)."""
    a=np.asarray(images,dtype=np.float32).reshape(-1,28,28).copy()
    kept=0;changed=0
    for j,im in enumerate(a):
        # Deterministic parity independent of label and class.
        proposed=cv2.dilate(im,KERNEL) if j%2 else cv2.erode(im,KERNEL)
        mixed=np.clip(.72*im+.28*proposed,0,255)
        if topology(mixed)==topology(im):
            a[j]=mixed
            changed+=1
        else:
            kept+=1
    return a.reshape(-1,784), {"stroke_changed":changed,"stroke_rejected_topology":kept}

def make_training(raw,y,target,variant):
    mode="size_com" if variant=="sizecom_balanced_scale" else "com"
    base=process(raw,mode,target)
    ids,targets=balanced_indices(y)
    augmented=np.empty((len(ids),784),dtype=np.float32)
    for desired in (16,18,20,22,24):
        mask=targets[ids]==desired
        s=ids[mask]
        assert len(s)>0 and len(s)==sum((np.sum(y==cl)//10) for cl in range(16))
        aug,_=apply_size(raw[s],desired)
        augmented[mask]=aug
    stroke_stats={"stroke_changed":0,"stroke_rejected_topology":0}
    if variant=="balanced_scale_stroke":
        augmented,stroke_stats=stroke_jitter(augmented,ids)
    # All evaluation uses the exact same deterministic mode, without augmentation.
    replacement=process(augmented,mode,target)
    base[ids]=replacement
    geom=image_geometry(base)
    sizes=[{"family":family,"n":int(np.sum(mask)),
             "longside_median":float(np.median(geom["long_side"][mask])),
             "longside_mean":float(np.mean(geom["long_side"][mask])),
             "mass_median":float(np.median(geom["mass"][mask])),
             "width_median":float(np.median(geom["width"][mask])),
             "height_median":float(np.median(geom["height"][mask]))}
            for family,mask in (("digits",y<10),("letters",y>=10))]
    return base,mode,stroke_stats,sizes

def group_metrics(y,pred,writers=None):
    masks={"all":np.ones(len(y),bool),"digits":y<10,"letters":y>=10}
    for j,l in enumerate(LABELS): masks["class_"+l]=y==j
    if writers is not None:
        for w in WRITERS:
            masks[f"writer_{w}"]=writers==w
            masks[f"writer_{w}_digits"]=(writers==w)&(y<10)
            masks[f"writer_{w}_letters"]=(writers==w)&(y>=10)
    return [{"group":name,"n":int(mask.sum()),"correct":int((pred[mask]==y[mask]).sum()),
             "accuracy":float(np.mean(pred[mask]==y[mask])),
             "macro_f1":float(f1_score(y[mask],pred[mask],
                 labels=np.unique(y[mask]),average="macro",zero_division=0))}
            for name,mask in masks.items() if mask.any()]

def original_model(path,seed,method,official,hm):
    d=Path(path)/f"normal-exp006-{method}-{seed}"
    manifest=json.loads((d/"run_manifest.json").read_text(encoding="utf-8"))
    assert manifest["method"]==method and manifest["seed"]==seed
    assert manifest["official_sha256"]==official["sha256"]
    assert all(manifest["writer_sha256"][str(w)]==hm[w]["sha256"] for w in WRITERS)
    assert sha256(d/"model.joblib")==manifest["model_sha256"]
    assert not manifest["official_test_used"] and not manifest["professor_test_used"]
    return joblib.load(d/"model.joblib"),manifest

def run(seed,model_root,out):
    out=Path(out)
    if out.exists(): raise FileExistsError(out)
    out.mkdir(parents=True)
    t0=time.monotonic()
    official=load_official(ROOT/"02_data"/"datasets"/"MNIST_EMNIST_official_split_112000.npz")
    hm=load_writers(ROOT/"02_data"/"Hand_made")
    Xraw=official["X_train"];ytr=np.asarray(official["y_train"],dtype=int)
    Xval=official["X_val"];yval=np.asarray(official["y_val"],dtype=int)
    Xhm=np.concatenate([hm[w]["X"] for w in WRITERS])
    yhm=np.concatenate([hm[w]["y"] for w in WRITERS]).astype(int)
    writers=np.repeat(WRITERS,640)
    assert Xraw.shape==(76800,784) and Xval.shape==(19200,784) and Xhm.shape==(2560,784)
    target=int(np.median(image_geometry(Xraw)["long_side"]))
    assert target==20
    models={};modes={};manifests={}
    for method in ("com","size_com"):
        m,h=original_model(model_root,seed,method,official,hm)
        name="baseline_"+method
        models[name]=m;modes[name]=method
        # Verify the archived official validation and handmade predictions.
        for dataset,X,y,record in (("OfficialVal",Xval,yval,h["official_validation"]),
                                    ("Handmade",Xhm,yhm,h["handmade"])):
            p=m.predict(process(X,method,target))
            for group,mask in (("all",np.ones(len(y),bool)),("digits",y<10),("letters",y>=10)):
                if int((p[mask]==y[mask]).sum())!=record[group]["correct"]:
                    raise AssertionError("Original EXP006 reproducibility failure: "+name+dataset+group)
        manifests[name]=h["model_sha256"]
    fitted_geometry=[]; fit_records=[]
    for variant in VARIANTS:
        xtr,mode,stroke_stat,geos=make_training(Xraw,ytr,target,variant)
        assert len(xtr)==76800
        estimator=Pipeline([("scale_0_1",PixelScaler255()),
                             ("model",MLPClassifier(random_state=seed))])
        tfit=time.monotonic()
        with warnings.catch_warnings(record=True) as seen:
            warnings.simplefilter("always")
            estimator.fit(xtr,ytr)
        sec=time.monotonic()-tfit
        models[variant]=estimator;modes[variant]=mode
        for v in geos: fitted_geometry.append({"variant":variant,**v})
        fit_records.append({"variant":variant,"seed":seed,"training_rows":len(xtr),
                            "fit_seconds":sec,"iterations":int(estimator.named_steps["model"].n_iter_),
                            "fit_warnings":" | ".join(str(x.message) for x in seen),
                            **stroke_stat})
        joblib.dump(estimator,out/f"{variant}.joblib",compress=3)
        print("EXP014_TRAINED",seed,variant,"seconds",round(sec,1),flush=True)
        del xtr
    save_csv(out/"training_geometry.csv",fitted_geometry)
    save_csv(out/"fit_records.csv",fit_records)
    metrics=[];pairs=[]
    for name,model in models.items():
        mode=modes[name]
        for dataset,X,y,w in (("OfficialVal",Xval,yval,None),("Handmade",Xhm,yhm,writers)):
            pred=np.asarray(model.predict(process(X,mode,target)),dtype=int)
            for gm in group_metrics(y,pred,w):
                metrics.append({"seed":seed,"model":name,"preprocessing":mode,
                                "dataset":dataset,**gm})
            if dataset=="Handmade":
                for a,b in PAIRS:
                    m=y==LABELS.index(a)
                    pairs.append({"seed":seed,"model":name,"actual":a,"predicted":b,
                                  "count":int(np.count_nonzero(pred[m]==LABELS.index(b)))})
    save_csv(out/"accuracy_metrics.csv",metrics)
    save_csv(out/"pair_errors.csv",pairs)
    scale_rows=[]; image_rows=[]
    for scale in SCALES:
        scaled=fixed_scale(Xhm,scale)
        features={mode:process(scaled,mode,target) for mode in ("com","size_com")}
        for name,model in models.items():
            mode=modes[name]
            proba=np.asarray(model.predict_proba(features[mode]),dtype=np.float64)
            labels=np.asarray(model.named_steps["model"].classes_,dtype=int)
            assert np.array_equal(labels,np.arange(16))
            pred=proba.argmax(axis=1)
            p_letter=proba[:,10:].sum(axis=1)
            for gm in group_metrics(yhm,pred,writers):
                scale_rows.append({"seed":seed,"model":name,"scale":scale,**gm,
                    "mean_p_letter":float(p_letter.mean())})
            for i in range(len(yhm)):
                image_rows.append({"seed":seed,"model":name,"scale":scale,
                    "writer":int(writers[i]),"index":int(i%640),
                    "true":LABELS[int(yhm[i])],"pred":LABELS[int(pred[i])],
                    "p_letter":float(p_letter[i])})
        print("EXP014_SCALED",seed,scale,flush=True)
    save_csv(out/"scale_metrics.csv",scale_rows)
    save_csv(out/"scale_image_family_probabilities.csv",image_rows)
    man={"experiment":"EXP014","seed":seed,"variants":list(VARIANTS),
         "scale_interventions":list(SCALES),"augmentation_fraction":.5,
         "augmented_targets_longside":[16,18,20,22,24],
         "balanced_targets_per_class":True,"training_rows_per_variant":76800,
         "model_hyperparameters":"sklearn MLPClassifier defaults, seed controlled",
         "archived_control_model_sha256":manifests,
         "original_checkpoint_hash_and_predictions_verified":True,
         "official_sha256":official["sha256"],
         "writer_sha256":{str(w):hm[w]["sha256"] for w in WRITERS},
         "original_test_used":False,"professor_test_used":False,
         "handmade_used_in_fit":False,"validation_used_in_fit":False,
         "training_time_seconds":time.monotonic()-t0}
    (out/"manifest.json").write_text(json.dumps(man,ensure_ascii=False,indent=2),encoding="utf-8")
    print("EXP014_COMPLETE",seed,flush=True)
if __name__=="__main__":
    p=argparse.ArgumentParser()
    p.add_argument("--seed",type=int,required=True,choices=(42,43,44))
    p.add_argument("--models-root",type=Path,required=True)
    p.add_argument("--out",type=Path,required=True)
    a=p.parse_args()
    run(a.seed,a.models_root,a.out)
```

## exp015_joint_rank

원본: [05_model_optimize/MLP/Normal/scripts/exp015_joint_rank.py](https://github.com/ryu980920/Machine_Learning/blob/b5810b1e677fdd4588b0c20df480aa83168df938/05_model_optimize/MLP/Normal/scripts/exp015_joint_rank.py)

EXP015 JOINT 45+33 run search: same preregistered size-shortcut safeguards.

Three independent MLP random initializations (seeds 42/43/44); the original

Gate established BEFORE model results; no post hoc relaxing of thresholds.

CLI 인자 (정적 추출): `--inputs`, `--out`

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp015_joint_rank.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""EXP015 JOINT 45+33 run search: same preregistered size-shortcut safeguards.

Three independent MLP random initializations (seeds 42/43/44); the original
Official Train, Official Val & Handmade datasets/splits are immutable.
Gate established BEFORE model results; no post hoc relaxing of thresholds.
"""
from __future__ import annotations
import argparse,csv,json,statistics
from pathlib import Path
from collections import defaultdict
import numpy as np
from exp006_size_com_comparison import LABELS

SEEDS=(42,43,44)
PCTS=(10,20,25,30,35,40,50)
STROKES=(0,14,28)
# These are PREDECLARED from EXP014's successful, conservative mean responses.
GATE={
 "digit_1p2_flip_mean_max":0.10,
 "letter_0p8_flip_mean_max":0.10,
 "digit_1p2_probability_drift_mean_max":0.12,
 "letter_0p8_probability_drift_mean_max":0.12,
 "each_seed_critical_flip_max":0.15,
 "all_scale_family_flip_mean_max":0.12,
 "handmade_digits_mean_min":0.85,
 "handmade_letters_mean_min":0.80,
}
def read(path):
    with Path(path).open("r",encoding="utf-8-sig",newline="") as f:
        return list(csv.DictReader(f))
def save(path,rows):
    if not rows:raise AssertionError(f"No data {path}")
    with Path(path).open("w",encoding="utf-8-sig",newline="") as f:
        w=csv.DictWriter(f,fieldnames=list(rows[0]))
        w.writeheader();w.writerows(rows)
def mean(values):
    assert values
    return float(sum(values)/len(values))
def run(inp,out):
    inp=Path(inp);out=Path(out)
    if out.exists():raise FileExistsError(out)
    out.mkdir(parents=True)
    folders=sorted(f for f in inp.iterdir() if f.is_dir() and
           (f.name.startswith("exp015-s") or f.name.startswith("exp015fine-s")))
    assert len(folders)==24,("expected 15 coarse + 9 fine jobs",len(folders))
    for f in folders:
        assert (f/"manifest.json").exists(), f
        m=json.loads((f/"manifest.json").read_text(encoding="utf-8"))
        assert m["seed"] in SEEDS and int(round(100*m["ratio_requested"])) in PCTS
        assert m["original_predictions_verified"]
        assert m["balanced_per_class_per_target"] and m["nested_sampling_from_exp014_half"]
        assert m["actual_training_rows_per_variant"]==76800
        assert not m["official_test_used"] and not m["professor_test_used"]
        assert not m["handmade_used_in_fit"] and not m["official_val_used_in_fit"]
    metrics=[r for f in folders for r in read(f/"accuracy_metrics.csv")]
    sensitivity=[r for f in folders for r in read(f/"scale_sensitivity.csv")]
    training=[r for f in folders for r in read(f/"training_records.csv")]
    geometry=[r for f in folders for r in read(f/"training_geometry.csv")]
    pair=[r for f in folders for r in read(f/"pair_errors.csv")]
    save(out/"all_seed_accuracy_metrics.csv",metrics)
    save(out/"all_seed_scale_sensitivity.csv",sensitivity)
    save(out/"all_seed_training_records.csv",training)
    save(out/"all_seed_training_geometry.csv",geometry)
    save(out/"all_seed_pair_errors.csv",pair)
    names=sorted({r["variant"] for r in training})
    assert len(names)==26,("expected 15 coarse + 11 new fine variants",len(names))
    source_by_model_seed={}
    for f in folders:
        for record in read(f/"training_records.csv"):
            key=(record["variant"],int(record["seed"]))
            assert key not in source_by_model_seed,("duplicate model run",key)
            source_by_model_seed[key]=f
    assert len(source_by_model_seed)==26*3
    lookup_acc={}
    for r in metrics:
        if r["group"] not in ("all","digits","letters"):continue
        key=(r["variant"],int(r["seed"]),r["dataset"],r["group"])
        if key in lookup_acc:
            # Baseline repeats in ratio==10? only appears in the 10-percent runs.
            raise AssertionError(("duplicate",key))
        lookup_acc[key]=float(r["accuracy"])
    lookup_robust={}
    for r in sensitivity:
        if r["group"] not in ("digits","letters"):continue
        key=(r["variant"],int(r["seed"]),r["group"],float(r["scale"]))
        assert key not in lookup_robust,key
        lookup_robust[key]=r
    lookup_train={}
    for r in training:
        key=(r["variant"],int(r["seed"]))
        assert key not in lookup_train,key
        lookup_train[key]=r
    allrows=[]
    for name in names:
        parts=name.split("_")
        ratio=int(parts[0][4:])/100.
        stroke=int(parts[1][6:])/100.
        def scores(dataset,group):
            return [lookup_acc[name,s,dataset,group] for s in SEEDS]
        val=scores("OfficialVal","all")
        digits=scores("Handmade","digits")
        letters=scores("Handmade","letters")
        def q(group,scale,field):
            return [float(lookup_robust[name,s,group,scale][field]) for s in SEEDS]
        d12=q("digits",1.2,"family_flip_rate")
        l08=q("letters",.8,"family_flip_rate")
        d12p=q("digits",1.2,"mean_abs_delta_p_letter")
        l08p=q("letters",.8,"mean_abs_delta_p_letter")
        allmeanflip=max(mean(q(group,scale,"family_flip_rate"))
            for group in ("digits","letters") for scale in (.8,.9,1.1,1.2))
        reasons=[]
        if mean(d12)>GATE["digit_1p2_flip_mean_max"]:reasons.append("digit_1p2_flip_mean")
        if mean(l08)>GATE["letter_0p8_flip_mean_max"]:reasons.append("letter_0p8_flip_mean")
        if mean(d12p)>GATE["digit_1p2_probability_drift_mean_max"]:reasons.append("digit_1p2_p_drift")
        if mean(l08p)>GATE["letter_0p8_probability_drift_mean_max"]:reasons.append("letter_0p8_p_drift")
        if max(d12+l08)>GATE["each_seed_critical_flip_max"]:reasons.append("per_seed_flip")
        if allmeanflip>GATE["all_scale_family_flip_mean_max"]:reasons.append("full_range_flip")
        if mean(digits)<GATE["handmade_digits_mean_min"]:reasons.append("digits_accuracy")
        if mean(letters)<GATE["handmade_letters_mean_min"]:reasons.append("letters_accuracy")
        train_sizes=[lookup_train[name,s]["size_only_val_roc_auc"] for s in SEEDS]
        assert len(set(train_sizes))==1,"Augmented geometry should not depend on MLP seed"
        record={"variant":name,"augmentation_fraction":ratio,"stroke_strength":stroke,
            "pass_shortcut_and_accuracy_gates":not reasons,"failed_gates":" | ".join(reasons),
            "official_val_mean":mean(val),"official_val_seed_sd":statistics.stdev(val),
            "handmade_digits_mean":mean(digits),"handmade_letters_mean":mean(letters),
            "handmade_min_family_mean":min(mean(digits),mean(letters)),
            "handmade_all_mean":mean(scores("Handmade","all")),
            "digits_1p2_flip_mean":mean(d12),"letters_0p8_flip_mean":mean(l08),
            "digits_1p2_mean_abs_probability_drift":mean(d12p),
            "letters_0p8_mean_abs_probability_drift":mean(l08p),
            "critical_max_seed_flip":max(d12+l08),
            "all_scale_max_family_flip_mean":allmeanflip,
            "train_geometry_only_size_auc_on_official_val":float(train_sizes[0]),
            "official_val_seed42":val[0],"official_val_seed43":val[1],
            "official_val_seed44":val[2],
            "handmade_digits_seed42":digits[0],"handmade_digits_seed43":digits[1],
            "handmade_digits_seed44":digits[2],
            "handmade_letters_seed42":letters[0],"handmade_letters_seed43":letters[1],
            "handmade_letters_seed44":letters[2]}
        allrows.append(record)
    # Strict binary eligibility gate, THEN rank OfficialVal, THEN worst family.
    winners=sorted((r for r in allrows if r["pass_shortcut_and_accuracy_gates"]),
                   key=lambda x:(-x["official_val_mean"],-x["handmade_min_family_mean"],
                                 x["all_scale_max_family_flip_mean"]))
    failures=sorted((r for r in allrows if not r["pass_shortcut_and_accuracy_gates"]),
                    key=lambda x:(len(x["failed_gates"].split(" | ")),-x["official_val_mean"]))
    ordered=winners+failures
    for i,r in enumerate(ordered):
        r["rank_among_eligible"]=i+1 if i<len(winners) else ""
        r["selection_status"]="ELIGIBLE" if i<len(winners) else "INELIGIBLE"
    save(out/"preregistered_gate_and_ranking.csv",ordered)
    # Additional class-specific mean family-flip of the direct paired counterfactual.
    class_rows=[]
    for name in names:
        ratio=int(name.split("_")[0][4:])
        for scale in (.8,1.2):
            for cl in ("0","8","5","B","D","F"):
                seed_flips=[];seed_abs=[]
                for seed in SEEDS:
                    f=source_by_model_seed[name,seed]
                    data=np.load(f/f"response_{name}.npz",allow_pickle=False)
                    ss=data["scales"].astype(np.float64)
                    ix=int(np.argmin(np.abs(ss-scale)));base=int(np.argmin(np.abs(ss-1.)))
                    p=data["p_letter"];lbl=data["y_true"]
                    mask=lbl==LABELS.index(cl)
                    assert np.sum(mask)==160 and abs(float(ss[ix])-scale)<1e-6
                    seed_flips.append(float(np.mean((p[ix,mask]>=.5)!=(p[base,mask]>=.5))))
                    seed_abs.append(float(np.mean(np.abs(p[ix,mask]-p[base,mask]))))
                class_rows.append({"variant":name,"class":cl,"scale":scale,
                    "n_unique_images_per_seed":160,
                    "family_flip_mean_over_seeds":mean(seed_flips),
                    "mean_abs_delta_p_letter":mean(seed_abs)})
    save(out/"class_specific_size_flip.csv",class_rows)
    # The frozen archive control is measured in the 10% jobs only.
    baseline={"OfficialVal":{},"Handmade":{}}
    for ds in baseline:
        for group in ("all","digits","letters"):
            scores=[lookup_acc["baseline_com",s,ds,group] for s in SEEDS]
            baseline[ds][group]={"mean":mean(scores),"stdev":statistics.stdev(scores)}
    control={}
    for group,scale in (("digits",1.2),("letters",.8)):
        control[group+"_"+str(scale)]={
          "mean_family_flip":mean(
               [float(lookup_robust["baseline_com",s,group,scale]["family_flip_rate"])
                for s in SEEDS]),
          "mean_abs_delta_p_letter":mean(
               [float(lookup_robust["baseline_com",s,group,scale]["mean_abs_delta_p_letter"])
                for s in SEEDS])}
    (out/"metadata.json").write_text(json.dumps({
        "experiment":"EXP015","three_seeds":list(SEEDS),"total_grid_conditions":len(names),
        "all_conditions_fit_on_original_train_only":True,"gate":GATE,
        "eligible_count":len(winners),"selected_best":winners[0]["variant"] if winners else None,
        "fixed_original_exp006_com_accuracy":baseline,"fixed_original_exp006_com_scale":control,
        "original_test_used":False,"professor_test_used":False,
        "main_or_final_checkpoint_changed":False
    },ensure_ascii=False,indent=2),encoding="utf-8")
    md=["# EXP유015 — 완전 검증된 3-Seed Grid 결과","",
        "**주의: 수치는 완료된 GitHub Actions Artifact의 실제 결과다.**",
        "Original Test 및 교수님 Test 미사용, Hand-made는 재사용 개발용 데이터.",
        f"후보 수: **{len(names)}**, Shortcut·숫자/문자 Accuracy 사전 게이트 통과: **{len(winners)}**.",
        "",
        "## 사전에 고정한 Shortcut 안전 게이트",
        "- 1.2배 숫자 및 0.8배 문자 Family flip 평균 각각 **10% 이하**",
        "- 같은 개입의 평균 절대 문자 확률 변화 각각 **0.12 이하**",
        "- 세 Seed 중 어느 한 Seed도 핵심 개입 Family flip이 **15% 초과 불가**",
        "- 0.8·0.9·1.1·1.2배의 모든 숫자/문자 집단에서 평균 flip **12% 이하**",
        "- 기본 Hand-made 숫자 **85% 이상**, 문자 **80% 이상** (각 3 Seed 평균)",
        "위 조건을 **모두 통과한 경우에만** OfficialVal 평균이 높은 순으로 비교한다.",
        "",
        "## 모든 후보",
        "| 조건 | 사전 Gate | OfficialVal | Handmade 숫자 | Handmade 문자 | 숫자×1.2 Flip | 문자×0.8 Flip |",
        "|---|---|---:|---:|---:|---:|---:|"]
    for row in ordered:
        md.append(f"| {row['variant']} | {'통과' if row['pass_shortcut_and_accuracy_gates'] else '탈락'} | "
           f"{100*row['official_val_mean']:.3f}% | "
           f"{100*row['handmade_digits_mean']:.3f}% | "
           f"{100*row['handmade_letters_mean']:.3f}% | "
           f"{100*row['digits_1p2_flip_mean']:.2f}% | "
           f"{100*row['letters_0p8_flip_mean']:.2f}% |")
    md+=["","## 사전 선정 규칙에 따른 결론"]
    if winners:
        best=winners[0]
        md += [f"- 게이트 통과 최상위 후보: **{best['variant']}**.",
               f"- OfficialVal: **{100*best['official_val_mean']:.4f}%**; "
               f"Handmade 숫자 **{100*best['handmade_digits_mean']:.4f}%**, "
               f"문자 **{100*best['handmade_letters_mean']:.4f}%**.",
               f"- 숫자×1.2 Family flip **{100*best['digits_1p2_flip_mean']:.3f}%**, "
               f"문자×0.8 Family flip **{100*best['letters_0p8_flip_mean']:.3f}%**.",
               f"- OfficialVal 97.30% 목표 {'충족' if best['official_val_mean']>=.973 else '미달'}."]
    else:
        md+=["- **통과 후보 없음:** Shortcut 억제 기준을 완화하거나 실패 후보를 임의 채택하지 않음."]
    md+=["",
      "### 해석상 주의",
      "- Family flip 감소는 크기 Shortcut 의존성 *완화*의 행동 근거이지 완전 제거의 인과적 증명은 아니다.",
      "- CoM-only 입력 전처리와 Augmentation 학습만 바뀌며 평가 입력에 증강 적용하지 않음.",
      "- OfficialVal과 Hand-made는 이미 반복 사용되었기 때문에 이 3-seed Grid의 최고 조건은 개발용 최적화 결과다. 새로운 작성자의 독립 평가 전까지 일반화 보증은 아님.",
      "- 원본 데이터와 main/최종 모델을 변경하지 않는다. 승인 전에 새 후보를 배포·적용하지 않는다.",
      ""]
    (out/"SUMMARY.md").write_text("\n".join(md),encoding="utf-8")
    print("EXP015_AGGREGATE_SUCCESS",len(names),"eligible",len(winners),
          "winner",winners[0]["variant"] if winners else None,flush=True)

if __name__=="__main__":
    p=argparse.ArgumentParser();p.add_argument("--inputs",type=Path,required=True)
    p.add_argument("--out",type=Path,required=True)
    a=p.parse_args();run(a.inputs,a.out)
```

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp015_scale_stroke_grid.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""EXP015: predeclared size-fraction x stroke-strength MLP training optimization.

All MLPs see EXACTLY 76,800 unique original Official Train source rows.
Handmade (2,560) & OfficialVal (19,200) are NEVER used in fitting or fitting augmentation.
No original Test or professor Test data access. Archived EXP006 control hash and predictions
must match before results are accepted. No mutations to final pipeline or main.
"""
from __future__ import annotations
import argparse, csv, json, time, warnings
from pathlib import Path
import cv2
import joblib
import numpy as np
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, balanced_accuracy_score, roc_auc_score, f1_score
from sklearn.neural_network import MLPClassifier

from exp006_size_com_comparison import (
    ROOT, WRITERS, LABELS, PixelScaler255, load_official, load_writers,
    image_geometry, apply_size, process, save_csv, sha256)
from exp011_012_causal_probes import fixed_scale
from exp014_balanced_augmentation import balanced_indices, original_model, group_metrics, PAIRS
from stroke_normalization import topology, KERNEL

SCALES=(.8,.9,1.0,1.1,1.2)
TARGETS=(16,18,20,22,24)
STROKES=(0.,.14,.28)  # same gray-level morphological blend as EXP014, varied strength
FRACTIONS=(.1,.2,.3,.4,.5)
SEEDS=(42,43,44)
AUG_SEED=20261116

def nested_subset(y,ratio):
    """Subsets of EXACT historic EXP014 50% IDs for controlled paired ratios.

    For each true class and each target size the chosen original source IDs
    are nested across fractions, sampled with reproducible independent RNG.
    At ratio .5 the IDs/size assignment are precisely EXP014 balanced_indices.
    """
    assert ratio in FRACTIONS
    ids50, target50=balanced_indices(y,seed=AUG_SEED)
    if ratio==.5: return ids50,target50
    chosen=[]
    for klass in range(16):
        for desired in TARGETS:
            source=ids50[(y[ids50]==klass)&(target50[ids50]==desired)]
            n=int(np.rint((ratio/.5)*len(source)))
            if not 0<n<=len(source): raise AssertionError((klass,desired,ratio))
            rng=np.random.default_rng(AUG_SEED+97*klass+desired*13)
            order=rng.permutation(len(source))
            # All fractions share the same order; smaller subsets nested in bigger.
            chosen.extend(source[order[:n]].tolist())
    ids=np.asarray(sorted(chosen),dtype=np.int32)
    assert len(np.unique(ids))==len(ids)
    assert abs(len(ids)/len(y)-ratio)<.003
    return ids,target50

def scale_base(raw,ids,targets):
    """Only original source bytes determine the augmentation; class independent size assignment."""
    resized=np.empty((len(ids),784),dtype=np.float32)
    for desired in TARGETS:
        at=np.flatnonzero(targets[ids]==desired)
        assert len(at)>0
        out,_=apply_size(raw[ids[at]],desired)
        resized[at]=out
    return resized

def stroke_jitter(images,strength):
    """Use identical morphological transform as EXP014, optional strength=0."""
    x=np.asarray(images,dtype=np.float32).reshape(-1,28,28).copy()
    if strength==0: return x.reshape(-1,784),{"modified":0,"rejected":0}
    changed=0;rejected=0
    for i,img in enumerate(x):
        target=cv2.erode(img,KERNEL) if (i%2)==0 else cv2.dilate(img,KERNEL)
        changed_img=np.clip((1-strength)*img+strength*target,0,255)
        if topology(changed_img)==topology(img):
            x[i]=changed_img
            changed+=1
        else:
            rejected+=1
    return x.reshape(-1,784),{"modified":changed,"rejected":rejected}

def preview_data(x):
    geo=image_geometry(x)
    return {k:np.asarray(geo[k]) for k in ("long_side","width","height","mass")}

def feature_probe(xtrain,ytrain,xval,yval):
    """Auxiliary diagnostic only: whether available size-only label signal remains."""
    train=preview_data(xtrain)["long_side"].reshape(-1,1)
    val=preview_data(xval)["long_side"].reshape(-1,1)
    yy=(ytrain>=10).astype(int)
    yv=(yval>=10).astype(int)
    m=make_pipeline(StandardScaler(),LogisticRegression(
        class_weight="balanced",random_state=42,max_iter=750))
    m.fit(train,yy)
    proba=m.predict_proba(val)[:,1]
    return {"size_only_val_bal_acc":float(balanced_accuracy_score(yv,proba>=.5)),
            "size_only_val_roc_auc":float(roc_auc_score(yv,proba))}

def summary_geometry(x,y):
    geo=preview_data(x)
    return [{"family":f,"count":int(m.sum()),
        **{k+"_mean":float(np.mean(v[m])) for k,v in geo.items()},
        **{k+"_median":float(np.median(v[m])) for k,v in geo.items()}}
        for f,m in (("digits",y<10),("letters",y>=10))]

def evaluate(model,seed,name,ratio,strength,xval,yval,xhm,yhm,writers,scaled, out):
    metrics=[];sensitivity=[];pair_errors=[]
    for ds,x,y,w in (("OfficialVal",xval,yval,None),("Handmade",xhm,yhm,writers)):
        prediction=np.asarray(model.predict(x),dtype=int)
        for r in group_metrics(y,prediction,w):
            metrics.append({"seed":seed,"variant":name,"ratio":ratio,"stroke_strength":strength,
                            "dataset":ds,**r})
        if ds=="Handmade":
            for a,b in PAIRS:
                m=y==LABELS.index(a)
                pair_errors.append({"seed":seed,"variant":name,"ratio":ratio,
                    "stroke_strength":strength,"actual":a,"predicted":b,
                    "count":int(np.sum(prediction[m]==LABELS.index(b)))})
    p=np.empty((len(SCALES),len(yhm)),dtype=np.float32)
    pred=np.empty((len(SCALES),len(yhm)),dtype=np.uint8)
    for k,scale in enumerate(SCALES):
        values=np.asarray(model.predict_proba(scaled[scale]),dtype=np.float32)
        labels=np.asarray(model.named_steps["model"].classes_,dtype=int)
        assert np.array_equal(labels,np.arange(16))
        p[k]=values[:,10:].sum(axis=1)
        pred[k]=values.argmax(axis=1)
    base_index=SCALES.index(1.0)
    for k,scale in enumerate(SCALES):
        for group,mask in (("digits",yhm<10),("letters",yhm>=10),
                           ("all",np.ones(len(yhm),bool)),
                           *[(f"writer_{w}_{family}",(writers==w)&mask)
                             for w in WRITERS for family,mask in
                             (("digits",yhm<10),("letters",yhm>=10))]):
            current=p[k,mask]
            base=p[base_index,mask]
            family_flip=(current>=.5)!=(base>=.5)
            accuracy=(pred[k,mask]==yhm[mask]).mean()
            sensitivity.append({"seed":seed,"variant":name,"ratio":ratio,
               "stroke_strength":strength,"scale":scale,"group":group,
               "n":int(mask.sum()),"family_flip_rate":float(np.mean(family_flip)),
               "mean_abs_delta_p_letter":float(np.mean(np.abs(current-base))),
               "mean_signed_delta_p_letter":float(np.mean(current-base)),
               "mean_p_letter":float(current.mean()),"accuracy":float(accuracy)})
    np.savez_compressed(out/f"response_{name}.npz",p_letter=p,cls_pred=pred,
        y_true=yhm.astype(np.uint8),writer=writers.astype(np.uint8),
        scales=np.asarray(SCALES,dtype=np.float32))
    return metrics,sensitivity,pair_errors

def run(seed,ratio,out,models_root):
    if seed not in SEEDS or ratio not in FRACTIONS: raise ValueError((seed,ratio))
    out=Path(out)
    if out.exists():raise FileExistsError(out)
    out.mkdir(parents=True)
    t0=time.monotonic()
    o=load_official(ROOT/"02_data"/"datasets"/"MNIST_EMNIST_official_split_112000.npz")
    hm=load_writers(ROOT/"02_data"/"Hand_made")
    Xraw=o["X_train"];ytrain=np.asarray(o["y_train"],dtype=int)
    Xval0=o["X_val"];yval=np.asarray(o["y_val"],dtype=int)
    Xhm0=np.concatenate([hm[w]["X"] for w in WRITERS])
    yhm=np.concatenate([hm[w]["y"] for w in WRITERS]).astype(int)
    writers=np.repeat(np.array(WRITERS),640)
    assert Xraw.shape==(76800,784) and Xval0.shape==(19200,784) and Xhm0.shape==(2560,784)
    target=int(np.median(image_geometry(Xraw)["long_side"]))
    assert target==20
    control,stored=original_model(models_root,seed,"com",o,hm)
    xval=process(Xval0,"com",target)
    xhm=process(Xhm0,"com",target)
    for X,y,which in ((xval,yval,"official_validation"),(xhm,yhm,"handmade")):
        cpred=np.asarray(control.predict(X),dtype=int)
        for group,mask in (("all",np.ones(len(y),bool)),("digits",y<10),("letters",y>=10)):
            assert int(np.sum(cpred[mask]==y[mask]))==stored[which][group]["correct"],(seed,which,group)
    scaled={s:process(fixed_scale(Xhm0,s),"com",target) for s in SCALES}
    base=process(Xraw,"com",target)
    ids,target_map=nested_subset(ytrain,ratio)
    resized=scale_base(Xraw,ids,target_map)
    assert len(base)==76800 and len(resized)==len(ids)
    expected_per_class_target={}
    for klass in range(16):
        counts=[int(np.sum((ytrain[ids]==klass)&(target_map[ids]==size))) for size in TARGETS]
        assert len(set(counts))==1,(klass,counts)
        expected_per_class_target[LABELS[klass]]=counts[0]
    print("EXP015_SETUP",seed,ratio,"n_aug",len(ids),"target counts",expected_per_class_target,flush=True)
    metrics=[];sensitivity=[];pairs=[];trains=[];geometry=[]
    if ratio==.1:   # archived CoM reference (duplicate if other ratios, skip)
        m,s,p=evaluate(control,seed,"baseline_com",0.,0.,xval,yval,xhm,yhm,writers,scaled,out)
        metrics+=m;sensitivity+=s;pairs+=p
    for stroke in STROKES:
        name=f"size{int(round(ratio*100)):02d}_stroke{int(round(stroke*100)):02d}"
        begin=time.monotonic()
        changed,stats=stroke_jitter(resized,stroke)
        xtrain=base.copy()
        xtrain[ids]=process(changed,"com",target)
        geom=summary_geometry(xtrain,ytrain)
        for g in geom:geometry.append({"seed":seed,"variant":name,"ratio":ratio,
            "stroke_strength":stroke,**g})
        probe=feature_probe(xtrain,ytrain,xval,yval)
        m=Pipeline([("scale_0_1",PixelScaler255()),
                    ("model",MLPClassifier(random_state=seed))])
        with warnings.catch_warnings(record=True) as notes:
            warnings.simplefilter("always")
            m.fit(xtrain,ytrain)
        fitted=time.monotonic()-begin
        joblib.dump(m,out/f"{name}.joblib",compress=3)
        tr={"seed":seed,"variant":name,"ratio":ratio,
             "actual_fraction":float(len(ids)/len(ytrain)),
             "stroke_strength":stroke,"n_train":int(len(ytrain)),
             "n_augment":int(len(ids)),"n_stroke_changed":stats["modified"],
             "n_stroke_rejected":stats["rejected"],
             "fit_time_seconds":fitted,
             "fit_iterations":int(m.named_steps["model"].n_iter_),
             "fit_warnings":" | ".join(str(x.message) for x in notes),
             **probe}
        trains.append(tr)
        a,b,c=evaluate(m,seed,name,ratio,stroke,xval,yval,xhm,yhm,writers,scaled,out)
        metrics+=a;sensitivity+=b;pairs+=c
        print("EXP015_MODEL",seed,name,"seconds",round(fitted,2),"eval",
              [(q["dataset"],q["group"],round(q["accuracy"],5)) for q in a
               if q["group"] in ("all","digits","letters")],flush=True)
        del changed,xtrain,m
    save_csv(out/"training_records.csv",trains)
    save_csv(out/"training_geometry.csv",geometry)
    save_csv(out/"accuracy_metrics.csv",metrics)
    save_csv(out/"scale_sensitivity.csv",sensitivity)
    save_csv(out/"pair_errors.csv",pairs)
    manifest={"experiment":"EXP015","seed":seed,"ratio_requested":ratio,
      "ratio_actual":len(ids)/len(ytrain),"stroke_strengths":list(STROKES),
      "scales":list(SCALES),"targets":list(TARGETS),
      "balanced_per_class_per_target":True,"nested_sampling_from_exp014_half":True,
      "official_train_n":76800,"official_val_n":19200,"handmade_n":2560,
      "official_source_sha256":o["sha256"],
      "handmade_source_sha256":{str(w):hm[w]["sha256"] for w in WRITERS},
      "archived_exp006_com_model_sha256":stored["model_sha256"],
      "original_predictions_verified":True,
      "actual_training_rows_per_variant":76800,"official_test_used":False,
      "professor_test_used":False,"handmade_used_in_fit":False,
      "official_val_used_in_fit":False,
      "total_run_seconds":time.monotonic()-t0}
    (out/"manifest.json").write_text(json.dumps(manifest,ensure_ascii=False,indent=2),encoding="utf-8")
    print("EXP015_COMPLETE",seed,ratio,round(time.monotonic()-t0,2),flush=True)

if __name__=="__main__":
    p=argparse.ArgumentParser()
    p.add_argument("--seed",type=int,required=True,choices=SEEDS)
    p.add_argument("--ratio-percent",type=int,required=True,choices=(10,20,30,40,50))
    p.add_argument("--models-root",type=Path,required=True)
    p.add_argument("--out",type=Path,required=True)
    a=p.parse_args()
    run(a.seed,a.ratio_percent/100,a.out,a.models_root)
```

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp016_high_fraction_grid.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""EXP016: extending class-balanced augmentation to 60-100 percent while preserving EXP015 50% anchor.

All MLPs see EXACTLY 76,800 unique original Official Train source rows.
Handmade (2,560) & OfficialVal (19,200) are NEVER used in fitting or fitting augmentation.
No original Test or professor Test data access. Archived EXP006 control hash and predictions
must match before results are accepted. No mutations to final pipeline or main.
"""
from __future__ import annotations
import argparse, csv, json, time, warnings
from pathlib import Path
import cv2
import joblib
import numpy as np
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, balanced_accuracy_score, roc_auc_score, f1_score
from sklearn.neural_network import MLPClassifier

from exp006_size_com_comparison import (
    ROOT, WRITERS, LABELS, PixelScaler255, load_official, load_writers,
    image_geometry, apply_size, process, save_csv, sha256)
from exp011_012_causal_probes import fixed_scale
from exp014_balanced_augmentation import balanced_indices, original_model, group_metrics, PAIRS
from stroke_normalization import topology, KERNEL

SCALES=(.8,.9,1.0,1.1,1.2)
TARGETS=(16,18,20,22,24)
STROKES=(0.,.14,.28,.42) # EXP015 intensity controls; fixed per training sample
FRACTIONS=(.6,.7,.8,.9,1.)
SEEDS=(42,43,44)
AUG_SEED=20261116

def nested_subset(y,ratio):
    """Nested 60-100% selection; exactly preserves preexisting EXP014/015 half.

    For each class:
      1. Retain all previously augmented EXP014 50%-subset IDs & size targets.
      2. Deterministically shuffle never-before-augmented remainder.
      3. Add 10% steps of original rows, using a balanced five-size cycle.
    Data source & target sizes are independent of MLP initialization seed.
    At ratio 100% every original training source row is present exactly once.
    """
    assert ratio in FRACTIONS
    half,half_target=balanced_indices(y,seed=AUG_SEED)
    targets=half_target.copy()
    selected=list(half.tolist())
    for klass in range(16):
        allcls=np.flatnonzero(y==klass)
        prev=half[y[half]==klass]
        rem=np.setdiff1d(allcls,prev,assume_unique=True)
        rng=np.random.default_rng(AUG_SEED+791*klass+29)
        rem=rng.permutation(rem)
        # Target sequence is balanced in every prefix; differences <=1 per class.
        # Use class-dependent rotation so first selected target is not class-favored.
        offset=(klass*3)%5
        extra_targets=np.array([TARGETS[(j+offset)%5] for j in range(len(rem))],dtype=np.int32)
        targets[rem]=extra_targets
        desired=int(np.rint(len(allcls)*ratio))
        howmany=desired-len(prev)
        assert 0<howmany<=len(rem),(klass,ratio,howmany,len(rem))
        selected.extend(rem[:howmany].tolist())
    ids=np.array(sorted(selected),dtype=np.int32)
    assert len(ids)==len(np.unique(ids)), "no duplicate train sources"
    assert np.isin(half,ids).all()
    assert abs(len(ids)/len(y)-ratio)<.002
    if ratio==1.0:assert len(ids)==len(y)
    for klass in range(16):
        counts=[int(np.sum((y[ids]==klass)&(targets[ids]==size))) for size in TARGETS]
        assert max(counts)-min(counts)<=1,(klass,ratio,counts)
    return ids,targets

def scale_base(raw,ids,targets):
    """Only original source bytes determine the augmentation; class independent size assignment."""
    resized=np.empty((len(ids),784),dtype=np.float32)
    for desired in TARGETS:
        at=np.flatnonzero(targets[ids]==desired)
        assert len(at)>0
        out,_=apply_size(raw[ids[at]],desired)
        resized[at]=out
    return resized

def stroke_jitter(images,strength,ids,ids_half):
    """Stable per-image morphology.

    For all historical 50% source IDs, stroke side matches EXP014/015
    "sorted half cohort, j%2" bit-for-bit. New source IDs use source index
    parity, independent of which upper fraction is selected.
    """
    x=np.asarray(images,dtype=np.float32).reshape(-1,28,28).copy()
    if strength==0:return x.reshape(-1,784),{"modified":0,"rejected":0}
    ids_half=np.asarray(ids_half,dtype=np.int32)
    half_rank=np.full(76800,-1,dtype=np.int32)
    half_rank[ids_half]=np.arange(len(ids_half),dtype=np.int32)
    parity=np.where(half_rank[ids]>=0,half_rank[ids]%2,ids%2)
    changed=0;rejected=0
    for j,img in enumerate(x):
        proposed=cv2.erode(img,KERNEL) if parity[j]==0 else cv2.dilate(img,KERNEL)
        mixed=np.clip((1-strength)*img+strength*proposed,0,255)
        if topology(mixed)==topology(img):
            x[j]=mixed
            changed+=1
        else:
            rejected+=1
    return x.reshape(-1,784),{"modified":changed,"rejected":rejected}

def preview_data(x):
    geo=image_geometry(x)
    return {k:np.asarray(geo[k]) for k in ("long_side","width","height","mass")}

def feature_probe(xtrain,ytrain,xval,yval):
    """Auxiliary diagnostic only: whether available size-only label signal remains."""
    train=preview_data(xtrain)["long_side"].reshape(-1,1)
    val=preview_data(xval)["long_side"].reshape(-1,1)
    yy=(ytrain>=10).astype(int)
    yv=(yval>=10).astype(int)
    m=make_pipeline(StandardScaler(),LogisticRegression(
        class_weight="balanced",random_state=42,max_iter=750))
    m.fit(train,yy)
    proba=m.predict_proba(val)[:,1]
    return {"size_only_val_bal_acc":float(balanced_accuracy_score(yv,proba>=.5)),
            "size_only_val_roc_auc":float(roc_auc_score(yv,proba))}

def summary_geometry(x,y):
    geo=preview_data(x)
    return [{"family":f,"count":int(m.sum()),
        **{k+"_mean":float(np.mean(v[m])) for k,v in geo.items()},
        **{k+"_median":float(np.median(v[m])) for k,v in geo.items()}}
        for f,m in (("digits",y<10),("letters",y>=10))]

def evaluate(model,seed,name,ratio,strength,xval,yval,xhm,yhm,writers,scaled, out):
    metrics=[];sensitivity=[];pair_errors=[]
    for ds,x,y,w in (("OfficialVal",xval,yval,None),("Handmade",xhm,yhm,writers)):
        prediction=np.asarray(model.predict(x),dtype=int)
        for r in group_metrics(y,prediction,w):
            metrics.append({"seed":seed,"variant":name,"ratio":ratio,"stroke_strength":strength,
                            "dataset":ds,**r})
        if ds=="Handmade":
            for a,b in PAIRS:
                m=y==LABELS.index(a)
                pair_errors.append({"seed":seed,"variant":name,"ratio":ratio,
                    "stroke_strength":strength,"actual":a,"predicted":b,
                    "count":int(np.sum(prediction[m]==LABELS.index(b)))})
    p=np.empty((len(SCALES),len(yhm)),dtype=np.float32)
    pred=np.empty((len(SCALES),len(yhm)),dtype=np.uint8)
    for k,scale in enumerate(SCALES):
        values=np.asarray(model.predict_proba(scaled[scale]),dtype=np.float32)
        labels=np.asarray(model.named_steps["model"].classes_,dtype=int)
        assert np.array_equal(labels,np.arange(16))
        p[k]=values[:,10:].sum(axis=1)
        pred[k]=values.argmax(axis=1)
    base_index=SCALES.index(1.0)
    for k,scale in enumerate(SCALES):
        for group,mask in (("digits",yhm<10),("letters",yhm>=10),
                           ("all",np.ones(len(yhm),bool)),
                           *[(f"writer_{w}_{family}",(writers==w)&mask)
                             for w in WRITERS for family,mask in
                             (("digits",yhm<10),("letters",yhm>=10))]):
            current=p[k,mask]
            base=p[base_index,mask]
            family_flip=(current>=.5)!=(base>=.5)
            accuracy=(pred[k,mask]==yhm[mask]).mean()
            sensitivity.append({"seed":seed,"variant":name,"ratio":ratio,
               "stroke_strength":strength,"scale":scale,"group":group,
               "n":int(mask.sum()),"family_flip_rate":float(np.mean(family_flip)),
               "mean_abs_delta_p_letter":float(np.mean(np.abs(current-base))),
               "mean_signed_delta_p_letter":float(np.mean(current-base)),
               "mean_p_letter":float(current.mean()),"accuracy":float(accuracy)})
    np.savez_compressed(out/f"response_{name}.npz",p_letter=p,cls_pred=pred,
        y_true=yhm.astype(np.uint8),writer=writers.astype(np.uint8),
        scales=np.asarray(SCALES,dtype=np.float32))
    return metrics,sensitivity,pair_errors

def run(seed,ratio,out,models_root):
    if seed not in SEEDS or ratio not in FRACTIONS: raise ValueError((seed,ratio))
    out=Path(out)
    if out.exists():raise FileExistsError(out)
    out.mkdir(parents=True)
    t0=time.monotonic()
    o=load_official(ROOT/"02_data"/"datasets"/"MNIST_EMNIST_official_split_112000.npz")
    hm=load_writers(ROOT/"02_data"/"Hand_made")
    Xraw=o["X_train"];ytrain=np.asarray(o["y_train"],dtype=int)
    Xval0=o["X_val"];yval=np.asarray(o["y_val"],dtype=int)
    Xhm0=np.concatenate([hm[w]["X"] for w in WRITERS])
    yhm=np.concatenate([hm[w]["y"] for w in WRITERS]).astype(int)
    writers=np.repeat(np.array(WRITERS),640)
    assert Xraw.shape==(76800,784) and Xval0.shape==(19200,784) and Xhm0.shape==(2560,784)
    target=int(np.median(image_geometry(Xraw)["long_side"]))
    assert target==20
    control,stored=original_model(models_root,seed,"com",o,hm)
    xval=process(Xval0,"com",target)
    xhm=process(Xhm0,"com",target)
    for X,y,which in ((xval,yval,"official_validation"),(xhm,yhm,"handmade")):
        cpred=np.asarray(control.predict(X),dtype=int)
        for group,mask in (("all",np.ones(len(y),bool)),("digits",y<10),("letters",y>=10)):
            assert int(np.sum(cpred[mask]==y[mask]))==stored[which][group]["correct"],(seed,which,group)
    scaled={s:process(fixed_scale(Xhm0,s),"com",target) for s in SCALES}
    base=process(Xraw,"com",target)
    ids,target_map=nested_subset(ytrain,ratio)
    ids50,_=balanced_indices(ytrain,seed=AUG_SEED)
    resized=scale_base(Xraw,ids,target_map)
    assert len(base)==76800 and len(resized)==len(ids)
    assert np.isin(ids50,ids).all(),"historical 50% images must be retained"
    expected_per_class_target={}
    for klass in range(16):
        counts=[int(np.sum((ytrain[ids]==klass)&(target_map[ids]==size))) for size in TARGETS]
        assert max(counts)-min(counts)<=1,(klass,counts)
        expected_per_class_target[LABELS[klass]]=counts
    print("EXP016_SETUP",seed,ratio,"n_aug",len(ids),"target counts",expected_per_class_target,flush=True)
    metrics=[];sensitivity=[];pairs=[];trains=[];geometry=[]
    # Frozen EXP006 baseline is verified above; EXP015 30/40/50 baselines are
    # compared in the separate aggregate job without retraining.
    for stroke in STROKES:
        name=f"size{int(round(ratio*100)):02d}_stroke{int(round(stroke*100)):02d}"
        begin=time.monotonic()
        changed,stats=stroke_jitter(resized,stroke,ids,ids50)
        xtrain=base.copy()
        xtrain[ids]=process(changed,"com",target)
        geom=summary_geometry(xtrain,ytrain)
        for g in geom:geometry.append({"seed":seed,"variant":name,"ratio":ratio,
            "stroke_strength":stroke,**g})
        probe=feature_probe(xtrain,ytrain,xval,yval)
        m=Pipeline([("scale_0_1",PixelScaler255()),
                    ("model",MLPClassifier(random_state=seed))])
        with warnings.catch_warnings(record=True) as notes:
            warnings.simplefilter("always")
            m.fit(xtrain,ytrain)
        fitted=time.monotonic()-begin
        joblib.dump(m,out/f"{name}.joblib",compress=3)
        tr={"seed":seed,"variant":name,"ratio":ratio,
             "actual_fraction":float(len(ids)/len(ytrain)),
             "stroke_strength":stroke,"n_train":int(len(ytrain)),
             "n_augment":int(len(ids)),"n_stroke_changed":stats["modified"],
             "n_stroke_rejected":stats["rejected"],
             "fit_time_seconds":fitted,
             "fit_iterations":int(m.named_steps["model"].n_iter_),
             "fit_warnings":" | ".join(str(x.message) for x in notes),
             **probe}
        trains.append(tr)
        a,b,c=evaluate(m,seed,name,ratio,stroke,xval,yval,xhm,yhm,writers,scaled,out)
        metrics+=a;sensitivity+=b;pairs+=c
        print("EXP016_MODEL",seed,name,"seconds",round(fitted,2),"eval",
              [(q["dataset"],q["group"],round(q["accuracy"],5)) for q in a
               if q["group"] in ("all","digits","letters")],flush=True)
        del changed,xtrain,m
    save_csv(out/"training_records.csv",trains)
    save_csv(out/"training_geometry.csv",geometry)
    save_csv(out/"accuracy_metrics.csv",metrics)
    save_csv(out/"scale_sensitivity.csv",sensitivity)
    save_csv(out/"pair_errors.csv",pairs)
    manifest={"experiment":"EXP016","seed":seed,"ratio_requested":ratio,
      "ratio_actual":len(ids)/len(ytrain),"stroke_strengths":list(STROKES),
      "scales":list(SCALES),"targets":list(TARGETS),
      "balanced_per_class_per_target":True,"nested_sampling_from_exp014_half":True,
      "augmentation_above_50_with_stable_stroke_parity":True,
      "official_train_n":76800,"official_val_n":19200,"handmade_n":2560,
      "official_source_sha256":o["sha256"],
      "handmade_source_sha256":{str(w):hm[w]["sha256"] for w in WRITERS},
      "archived_exp006_com_model_sha256":stored["model_sha256"],
      "original_predictions_verified":True,
      "actual_training_rows_per_variant":76800,"official_test_used":False,
      "professor_test_used":False,"handmade_used_in_fit":False,
      "official_val_used_in_fit":False,
      "total_run_seconds":time.monotonic()-t0}
    (out/"manifest.json").write_text(json.dumps(manifest,ensure_ascii=False,indent=2),encoding="utf-8")
    print("EXP016_COMPLETE",seed,ratio,round(time.monotonic()-t0,2),flush=True)

if __name__=="__main__":
    p=argparse.ArgumentParser()
    p.add_argument("--seed",type=int,required=True,choices=SEEDS)
    p.add_argument("--ratio-percent",type=int,required=True,choices=(60,70,80,90,100))
    p.add_argument("--models-root",type=Path,required=True)
    p.add_argument("--out",type=Path,required=True)
    a=p.parse_args()
    run(a.seed,a.ratio_percent/100,a.out,a.models_root)
```

## exp015_aggregate_and_rank

CLI 인자 (정적 추출): `--inputs`, `--out`

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp015_aggregate_and_rank.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""EXP015: rank candidates ONLY after preregistered size-shortcut safeguards.

Three independent MLP random initializations (seeds 42/43/44); the original
Official Train, Official Val & Handmade datasets/splits are immutable.
Gate established BEFORE model results; no post hoc relaxing of thresholds.
"""
from __future__ import annotations
import argparse,csv,json,statistics
from pathlib import Path
from collections import defaultdict
import numpy as np
from exp006_size_com_comparison import LABELS

SEEDS=(42,43,44)
PCTS=(10,20,30,40,50)
STROKES=(0,14,28)
# These are PREDECLARED from EXP014's successful, conservative mean responses.
GATE={
 "digit_1p2_flip_mean_max":0.10,
 "letter_0p8_flip_mean_max":0.10,
 "digit_1p2_probability_drift_mean_max":0.12,
 "letter_0p8_probability_drift_mean_max":0.12,
 "each_seed_critical_flip_max":0.15,
 "all_scale_family_flip_mean_max":0.12,
 "handmade_digits_mean_min":0.85,
 "handmade_letters_mean_min":0.80,
}
def read(path):
    with Path(path).open("r",encoding="utf-8-sig",newline="") as f:
        return list(csv.DictReader(f))
def save(path,rows):
    if not rows:raise AssertionError(f"No data {path}")
    with Path(path).open("w",encoding="utf-8-sig",newline="") as f:
        w=csv.DictWriter(f,fieldnames=list(rows[0]))
        w.writeheader();w.writerows(rows)
def mean(values):
    assert values
    return float(sum(values)/len(values))
def run(inp,out):
    inp=Path(inp);out=Path(out)
    if out.exists():raise FileExistsError(out)
    out.mkdir(parents=True)
    folders=[inp/f"exp015-s{s}-r{pct}" for s in SEEDS for pct in PCTS]
    for f in folders:
        assert (f/"manifest.json").exists(), f
        m=json.loads((f/"manifest.json").read_text(encoding="utf-8"))
        assert m["seed"] in SEEDS and int(round(100*m["ratio_requested"])) in PCTS
        assert m["original_predictions_verified"]
        assert m["balanced_per_class_per_target"] and m["nested_sampling_from_exp014_half"]
        assert m["actual_training_rows_per_variant"]==76800
        assert not m["official_test_used"] and not m["professor_test_used"]
        assert not m["handmade_used_in_fit"] and not m["official_val_used_in_fit"]
    metrics=[r for f in folders for r in read(f/"accuracy_metrics.csv")]
    sensitivity=[r for f in folders for r in read(f/"scale_sensitivity.csv")]
    training=[r for f in folders for r in read(f/"training_records.csv")]
    geometry=[r for f in folders for r in read(f/"training_geometry.csv")]
    pair=[r for f in folders for r in read(f/"pair_errors.csv")]
    save(out/"all_seed_accuracy_metrics.csv",metrics)
    save(out/"all_seed_scale_sensitivity.csv",sensitivity)
    save(out/"all_seed_training_records.csv",training)
    save(out/"all_seed_training_geometry.csv",geometry)
    save(out/"all_seed_pair_errors.csv",pair)
    names=sorted({r["variant"] for r in training})
    assert len(names)==len(PCTS)*len(STROKES),len(names)
    lookup_acc={}
    for r in metrics:
        if r["group"] not in ("all","digits","letters"):continue
        key=(r["variant"],int(r["seed"]),r["dataset"],r["group"])
        if key in lookup_acc:
            # Baseline repeats in ratio==10? only appears in the 10-percent runs.
            raise AssertionError(("duplicate",key))
        lookup_acc[key]=float(r["accuracy"])
    lookup_robust={}
    for r in sensitivity:
        if r["group"] not in ("digits","letters"):continue
        key=(r["variant"],int(r["seed"]),r["group"],float(r["scale"]))
        assert key not in lookup_robust,key
        lookup_robust[key]=r
    lookup_train={}
    for r in training:
        key=(r["variant"],int(r["seed"]))
        assert key not in lookup_train,key
        lookup_train[key]=r
    allrows=[]
    for name in names:
        parts=name.split("_")
        ratio=int(parts[0][4:])/100.
        stroke=int(parts[1][6:])/100.
        def scores(dataset,group):
            return [lookup_acc[name,s,dataset,group] for s in SEEDS]
        val=scores("OfficialVal","all")
        digits=scores("Handmade","digits")
        letters=scores("Handmade","letters")
        def q(group,scale,field):
            return [float(lookup_robust[name,s,group,scale][field]) for s in SEEDS]
        d12=q("digits",1.2,"family_flip_rate")
        l08=q("letters",.8,"family_flip_rate")
        d12p=q("digits",1.2,"mean_abs_delta_p_letter")
        l08p=q("letters",.8,"mean_abs_delta_p_letter")
        allmeanflip=max(mean(q(group,scale,"family_flip_rate"))
            for group in ("digits","letters") for scale in (.8,.9,1.1,1.2))
        reasons=[]
        if mean(d12)>GATE["digit_1p2_flip_mean_max"]:reasons.append("digit_1p2_flip_mean")
        if mean(l08)>GATE["letter_0p8_flip_mean_max"]:reasons.append("letter_0p8_flip_mean")
        if mean(d12p)>GATE["digit_1p2_probability_drift_mean_max"]:reasons.append("digit_1p2_p_drift")
        if mean(l08p)>GATE["letter_0p8_probability_drift_mean_max"]:reasons.append("letter_0p8_p_drift")
        if max(d12+l08)>GATE["each_seed_critical_flip_max"]:reasons.append("per_seed_flip")
        if allmeanflip>GATE["all_scale_family_flip_mean_max"]:reasons.append("full_range_flip")
        if mean(digits)<GATE["handmade_digits_mean_min"]:reasons.append("digits_accuracy")
        if mean(letters)<GATE["handmade_letters_mean_min"]:reasons.append("letters_accuracy")
        train_sizes=[lookup_train[name,s]["size_only_val_roc_auc"] for s in SEEDS]
        assert len(set(train_sizes))==1,"Augmented geometry should not depend on MLP seed"
        record={"variant":name,"augmentation_fraction":ratio,"stroke_strength":stroke,
            "pass_shortcut_and_accuracy_gates":not reasons,"failed_gates":" | ".join(reasons),
            "official_val_mean":mean(val),"official_val_seed_sd":statistics.stdev(val),
            "handmade_digits_mean":mean(digits),"handmade_letters_mean":mean(letters),
            "handmade_min_family_mean":min(mean(digits),mean(letters)),
            "handmade_all_mean":mean(scores("Handmade","all")),
            "digits_1p2_flip_mean":mean(d12),"letters_0p8_flip_mean":mean(l08),
            "digits_1p2_mean_abs_probability_drift":mean(d12p),
            "letters_0p8_mean_abs_probability_drift":mean(l08p),
            "critical_max_seed_flip":max(d12+l08),
            "all_scale_max_family_flip_mean":allmeanflip,
            "train_geometry_only_size_auc_on_official_val":float(train_sizes[0]),
            "official_val_seed42":val[0],"official_val_seed43":val[1],
            "official_val_seed44":val[2],
            "handmade_digits_seed42":digits[0],"handmade_digits_seed43":digits[1],
            "handmade_digits_seed44":digits[2],
            "handmade_letters_seed42":letters[0],"handmade_letters_seed43":letters[1],
            "handmade_letters_seed44":letters[2]}
        allrows.append(record)
    # Strict binary eligibility gate, THEN rank OfficialVal, THEN worst family.
    winners=sorted((r for r in allrows if r["pass_shortcut_and_accuracy_gates"]),
                   key=lambda x:(-x["official_val_mean"],-x["handmade_min_family_mean"],
                                 x["all_scale_max_family_flip_mean"]))
    failures=sorted((r for r in allrows if not r["pass_shortcut_and_accuracy_gates"]),
                    key=lambda x:(len(x["failed_gates"].split(" | ")),-x["official_val_mean"]))
    ordered=winners+failures
    for i,r in enumerate(ordered):
        r["rank_among_eligible"]=i+1 if i<len(winners) else ""
        r["selection_status"]="ELIGIBLE" if i<len(winners) else "INELIGIBLE"
    save(out/"preregistered_gate_and_ranking.csv",ordered)
    # Additional class-specific mean family-flip of the direct paired counterfactual.
    class_rows=[]
    for name in names:
        ratio=int(name.split("_")[0][4:])
        for scale in (.8,1.2):
            for cl in ("0","8","5","B","D","F"):
                seed_flips=[];seed_abs=[]
                for seed in SEEDS:
                    f=inp/f"exp015-s{seed}-r{ratio}"
                    data=np.load(f/f"response_{name}.npz",allow_pickle=False)
                    ss=data["scales"].astype(np.float64)
                    ix=int(np.argmin(np.abs(ss-scale)));base=int(np.argmin(np.abs(ss-1.)))
                    p=data["p_letter"];lbl=data["y_true"]
                    mask=lbl==LABELS.index(cl)
                    assert np.sum(mask)==160 and abs(float(ss[ix])-scale)<1e-6
                    seed_flips.append(float(np.mean((p[ix,mask]>=.5)!=(p[base,mask]>=.5))))
                    seed_abs.append(float(np.mean(np.abs(p[ix,mask]-p[base,mask]))))
                class_rows.append({"variant":name,"class":cl,"scale":scale,
                    "n_unique_images_per_seed":160,
                    "family_flip_mean_over_seeds":mean(seed_flips),
                    "mean_abs_delta_p_letter":mean(seed_abs)})
    save(out/"class_specific_size_flip.csv",class_rows)
    # The frozen archive control is measured in the 10% jobs only.
    baseline={"OfficialVal":{},"Handmade":{}}
    for ds in baseline:
        for group in ("all","digits","letters"):
            scores=[lookup_acc["baseline_com",s,ds,group] for s in SEEDS]
            baseline[ds][group]={"mean":mean(scores),"stdev":statistics.stdev(scores)}
    control={}
    for group,scale in (("digits",1.2),("letters",.8)):
        control[group+"_"+str(scale)]={
          "mean_family_flip":mean(
               [float(lookup_robust["baseline_com",s,group,scale]["family_flip_rate"])
                for s in SEEDS]),
          "mean_abs_delta_p_letter":mean(
               [float(lookup_robust["baseline_com",s,group,scale]["mean_abs_delta_p_letter"])
                for s in SEEDS])}
    (out/"metadata.json").write_text(json.dumps({
        "experiment":"EXP015","three_seeds":list(SEEDS),"total_grid_conditions":len(names),
        "all_conditions_fit_on_original_train_only":True,"gate":GATE,
        "eligible_count":len(winners),"selected_best":winners[0]["variant"] if winners else None,
        "fixed_original_exp006_com_accuracy":baseline,"fixed_original_exp006_com_scale":control,
        "original_test_used":False,"professor_test_used":False,
        "main_or_final_checkpoint_changed":False
    },ensure_ascii=False,indent=2),encoding="utf-8")
    md=["# EXP유015 — 완전 검증된 3-Seed Grid 결과","",
        "**주의: 수치는 완료된 GitHub Actions Artifact의 실제 결과다.**",
        "Original Test 및 교수님 Test 미사용, Hand-made는 재사용 개발용 데이터.",
        f"후보 수: **{len(names)}**, Shortcut·숫자/문자 Accuracy 사전 게이트 통과: **{len(winners)}**.",
        "",
        "## 사전에 고정한 Shortcut 안전 게이트",
        "- 1.2배 숫자 및 0.8배 문자 Family flip 평균 각각 **10% 이하**",
        "- 같은 개입의 평균 절대 문자 확률 변화 각각 **0.12 이하**",
        "- 세 Seed 중 어느 한 Seed도 핵심 개입 Family flip이 **15% 초과 불가**",
        "- 0.8·0.9·1.1·1.2배의 모든 숫자/문자 집단에서 평균 flip **12% 이하**",
        "- 기본 Hand-made 숫자 **85% 이상**, 문자 **80% 이상** (각 3 Seed 평균)",
        "위 조건을 **모두 통과한 경우에만** OfficialVal 평균이 높은 순으로 비교한다.",
        "",
        "## 모든 후보",
        "| 조건 | 사전 Gate | OfficialVal | Handmade 숫자 | Handmade 문자 | 숫자×1.2 Flip | 문자×0.8 Flip |",
        "|---|---|---:|---:|---:|---:|---:|"]
    for row in ordered:
        md.append(f"| {row['variant']} | {'통과' if row['pass_shortcut_and_accuracy_gates'] else '탈락'} | "
           f"{100*row['official_val_mean']:.3f}% | "
           f"{100*row['handmade_digits_mean']:.3f}% | "
           f"{100*row['handmade_letters_mean']:.3f}% | "
           f"{100*row['digits_1p2_flip_mean']:.2f}% | "
           f"{100*row['letters_0p8_flip_mean']:.2f}% |")
    md+=["","## 사전 선정 규칙에 따른 결론"]
    if winners:
        best=winners[0]
        md += [f"- 게이트 통과 최상위 후보: **{best['variant']}**.",
               f"- OfficialVal: **{100*best['official_val_mean']:.4f}%**; "
               f"Handmade 숫자 **{100*best['handmade_digits_mean']:.4f}%**, "
               f"문자 **{100*best['handmade_letters_mean']:.4f}%**.",
               f"- 숫자×1.2 Family flip **{100*best['digits_1p2_flip_mean']:.3f}%**, "
               f"문자×0.8 Family flip **{100*best['letters_0p8_flip_mean']:.3f}%**.",
               f"- OfficialVal 97.30% 목표 {'충족' if best['official_val_mean']>=.973 else '미달'}."]
    else:
        md+=["- **통과 후보 없음:** Shortcut 억제 기준을 완화하거나 실패 후보를 임의 채택하지 않음."]
    md+=["",
      "### 해석상 주의",
      "- Family flip 감소는 크기 Shortcut 의존성 *완화*의 행동 근거이지 완전 제거의 인과적 증명은 아니다.",
      "- CoM-only 입력 전처리와 Augmentation 학습만 바뀌며 평가 입력에 증강 적용하지 않음.",
      "- OfficialVal과 Hand-made는 이미 반복 사용되었기 때문에 이 3-seed Grid의 최고 조건은 개발용 최적화 결과다. 새로운 작성자의 독립 평가 전까지 일반화 보증은 아님.",
      "- 원본 데이터와 main/최종 모델을 변경하지 않는다. 승인 전에 새 후보를 배포·적용하지 않는다.",
      ""]
    (out/"SUMMARY.md").write_text("\n".join(md),encoding="utf-8")
    print("EXP015_AGGREGATE_SUCCESS",len(names),"eligible",len(winners),
          "winner",winners[0]["variant"] if winners else None,flush=True)

if __name__=="__main__":
    p=argparse.ArgumentParser();p.add_argument("--inputs",type=Path,required=True)
    p.add_argument("--out",type=Path,required=True)
    a=p.parse_args();run(a.inputs,a.out)
```

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp016_aggregate_shortcut_risks.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""Aggregate EXP016 60 high-ratio new MLPs vs frozen EXP015 30/40/50 reference MLPs.

Never retrain or read any Original Test/Professor Test data.
"""
from __future__ import annotations
import argparse,csv,json,statistics
from collections import defaultdict
from pathlib import Path
import numpy as np
from exp006_size_com_comparison import LABELS

SEEDS=(42,43,44)
HIGH=(60,70,80,90,100)
STROKES=(0,14,28,42)
OLD=(30,40,50)
FOCUS=("8","B","D")
SC=(.8,.9,1.,1.1,1.2)
EXTREMES=(.8,1.2)
GATES={"digit12flip":.10,"letter08flip":.10,"digit12p":.12,"letter08p":.12,
       "seedgroup":.15,"allscale":.12,"digitsacc":.85,"lettersacc":.80,
       "focusmax":.20,"allclassmax":.25,"seedfocusmax":.30}
EPS=1e-10
def read(path):
    with Path(path).open("r",encoding="utf-8-sig",newline="") as f:
        return list(csv.DictReader(f))
def save(path,rows):
    assert rows
    with Path(path).open("w",encoding="utf-8-sig",newline="") as f:
        w=csv.DictWriter(f,fieldnames=list(rows[0]));w.writeheader();w.writerows(rows)
def avg(x):return sum(x)/len(x)
def run(hi,old,out):
    hi=Path(hi);old=Path(old);out=Path(out)
    assert not out.exists()
    out.mkdir(parents=True)
    models_new={f"size{r}_stroke{v:02d}" for r in HIGH for v in STROKES}
    models_old={f"size{r}_stroke28" for r in OLD}
    names=sorted(models_new|models_old)
    assert len(names)==23
    record={};acc={};sens={};pairs=[];src={}
    for s in SEEDS:
        for high,ratios,root in ((True,HIGH,hi),(False,OLD,old)):
            for ratio in ratios:
                f=root/(f"exp016-s{s}-r{ratio}" if high else f"exp015-s{s}-r{ratio}")
                manifest=json.loads((f/"manifest.json").read_text(encoding="utf-8"))
                assert manifest["seed"]==s and manifest["actual_training_rows_per_variant"]==76800
                assert manifest["experiment"]==("EXP016" if high else "EXP015")
                assert manifest["original_predictions_verified"]
                assert manifest["official_train_n"]==76800 and manifest["official_val_n"]==19200
                assert manifest["handmade_n"]==2560
                assert manifest["nested_sampling_from_exp014_half"]
                assert not any(manifest[k] for k in ("official_test_used","professor_test_used",
                              "handmade_used_in_fit","official_val_used_in_fit"))
                if high:
                    assert manifest["augmentation_above_50_with_stable_stroke_parity"]
                    assert abs(manifest["ratio_actual"]-ratio/100)<.002
                desired={f"size{ratio}_stroke{v:02d}" for v in STROKES} if high else {f"size{ratio}_stroke28"}
                for row in read(f/"training_records.csv"):
                    if row["variant"] in desired:
                        key=(row["variant"],s)
                        assert key not in record,key
                        record[key]=row;src[key]=f
                for row in read(f/"accuracy_metrics.csv"):
                    if row["variant"] not in desired or row["group"] not in ("all","digits","letters"):continue
                    key=(row["variant"],s,row["dataset"],row["group"])
                    assert key not in acc,key
                    acc[key]=float(row["accuracy"])
                for row in read(f/"scale_sensitivity.csv"):
                    if row["variant"] not in desired:continue
                    key=(row["variant"],s,row["group"],float(row["scale"]))
                    assert key not in sens,key
                    sens[key]=row
                pairs += [x for x in read(f/"pair_errors.csv") if x["variant"] in desired]
    assert len(record)==69,len(record)
    assert all((name,s) in src for name in names for s in SEEDS)
    save(out/"all_candidates_seed_pair_errors.csv",pairs)
    cls_data={};perclass=[];seed_focus={}
    for name in names:
        scratch=defaultdict(list)
        for s in SEEDS:
            path=src[name,s]/f"response_{name}.npz"
            with np.load(path,allow_pickle=False) as z:
                p=np.array(z["p_letter"])
                truth=np.array(z["y_true"],dtype=int)
                scales=np.array(z["scales"],dtype=float)
            assert p.shape==(5,2560) and len(truth)==2560
            assert np.allclose(scales,SC,atol=1e-6)
            assert set(truth)==set(range(16))
            baseline=p[2]>=.5
            for c,label in enumerate(LABELS):
                mask=truth==c
                assert np.sum(mask)==160,(name,s,label)
                for k,scale in enumerate(SC):
                    if scale==1:continue
                    flip=float(np.mean((p[k,mask]>=.5)!=baseline[mask]))
                    drift=float(np.mean(np.abs(p[k,mask]-p[2,mask])))
                    scratch[label,scale].append((flip,drift))
                    if label in FOCUS and scale in EXTREMES:
                        key=(name,s)
                        seed_focus[key]=max(seed_focus.get(key,0.),flip)
        for (label,scale),values in sorted(scratch.items()):
            assert len(values)==3
            row={"model":name,"class":label,"group":"digits" if LABELS.index(label)<10 else "letters",
                "scale":scale,"n_per_seed":160,"seeds":3,
                "flip_mean":avg([a for a,b in values]),
                "flip_max_seed":max(a for a,b in values),
                "abs_delta_p_letter_mean":avg([b for a,b in values])}
            perclass.append(row);cls_data[name,label,scale]=row
    save(out/"all_16class_size_flip_3seed.csv",perclass)
    ranked=[]
    for name in names:
        def accuracy(ds,group):return [acc[name,s,ds,group] for s in SEEDS]
        def metric(group,scale,field):
            return [float(sens[name,s,group,scale][field]) for s in SEEDS]
        v=accuracy("OfficialVal","all")
        d=accuracy("Handmade","digits");l=accuracy("Handmade","letters")
        d12=metric("digits",1.2,"family_flip_rate")
        l08=metric("letters",.8,"family_flip_rate")
        dp=metric("digits",1.2,"mean_abs_delta_p_letter")
        lp=metric("letters",.8,"mean_abs_delta_p_letter")
        allgroup=max(avg(metric(g,sc,"family_flip_rate")) for g in ("digits","letters") for sc in (.8,.9,1.1,1.2))
        focusmax,cl,scale=max((cls_data[name,c,sc]["flip_mean"],c,sc) for c in FOCUS for sc in EXTREMES)
        anymax=max(cls_data[name,c,sc]["flip_mean"] for c in LABELS for sc in EXTREMES)
        seedsmax=max(seed_focus[name,s] for s in SEEDS)
        failed=[]
        if avg(d12)>GATES["digit12flip"]+EPS:failed.append("digit_1p2_flip")
        if avg(l08)>GATES["letter08flip"]+EPS:failed.append("letter_0p8_flip")
        if avg(dp)>GATES["digit12p"]+EPS:failed.append("digit_1p2_prob_delta")
        if avg(lp)>GATES["letter08p"]+EPS:failed.append("letter_0p8_prob_delta")
        if max(d12+l08)>GATES["seedgroup"]+EPS:failed.append("seed_group_flip")
        if allgroup>GATES["allscale"]+EPS:failed.append("all_scale_group_flip")
        if avg(d)<GATES["digitsacc"]-EPS:failed.append("digit_accuracy")
        if avg(l)<GATES["lettersacc"]-EPS:failed.append("letter_accuracy")
        old_ok=not failed
        if focusmax>GATES["focusmax"]+EPS:failed.append("B8D_worst_flip")
        if anymax>GATES["allclassmax"]+EPS:failed.append("any16class_worst_flip")
        if seedsmax>GATES["seedfocusmax"]+EPS:failed.append("B8D_worst_per_seed_flip")
        ranked.append({"variant":name,"cohort":"EXP016_NEW" if name in models_new else "EXP015_FROZEN",
             "size_augmentation_pct":int(name.split("_")[0][4:]),
             "stroke_blending_pct":int(name.split("_")[1][6:]),
             "passes_legacy_group_gates":old_ok,"passes_all_predeclared_gates":not failed,
             "failed_gates":" | ".join(failed),
             "official_val_mean":avg(v),"official_val_stdev":statistics.stdev(v),
             "official_val_seed42":v[0],"official_val_seed43":v[1],"official_val_seed44":v[2],
             "handmade_digits_mean":avg(d),"handmade_letters_mean":avg(l),
             "handmade_all_mean":avg(accuracy("Handmade","all")),
             "digits_1p2_flip":avg(d12),"letters_0p8_flip":avg(l08),
             "digits_1p2_mean_abs_delta_p":avg(dp),"letters_0p8_mean_abs_delta_p":avg(lp),
             "max_all_group_scale_flip":allgroup,
             "max_seed_group_flip":max(d12+l08),
             "B8D_max_class_flip":focusmax,"B8D_worst_class":cl,"B8D_worst_scale":scale,
             "all16_max_class_flip":anymax,"B8D_max_single_seed_class_flip":seedsmax,
             "size_only_geometry_bal_acc":float(record[name,42]["size_only_val_bal_acc"]),
             "size_only_geometry_auc":float(record[name,42]["size_only_val_roc_auc"]),
             "actual_augmented_fraction":float(record[name,42]["actual_fraction"])})
    eligible=sorted([x for x in ranked if x["passes_all_predeclared_gates"]],
       key=lambda x:(x["all16_max_class_flip"],x["B8D_max_class_flip"],
                     -x["official_val_mean"],-min(x["handmade_digits_mean"],x["handmade_letters_mean"])))
    rejected=sorted([x for x in ranked if not x["passes_all_predeclared_gates"]],
       key=lambda x:(x["all16_max_class_flip"],x["B8D_max_class_flip"],-x["official_val_mean"]))
    ordered=eligible+rejected
    for j,x in enumerate(ordered):
        x["eligible_rank"]=j+1 if j<len(eligible) else ""
        x["decision"]="ELIGIBLE" if j<len(eligible) else "INELIGIBLE"
    save(out/"all_23_conditions_ranked_gates.csv",ordered)
    best_focus=min(ranked,key=lambda x:(x["B8D_max_class_flip"],x["all16_max_class_flip"]))
    metadata={"experiment":"EXP016","seeds":list(SEEDS),"new_MLPs":60,"new_conditions":20,
         "historical_frozen_controls":sorted(models_old),"candidates":len(ranked),
         "legacy_gate_pass":sum(x["passes_legacy_group_gates"] for x in ranked),
         "strict_all_gates_pass":len(eligible),"strict_top_candidate":eligible[0]["variant"] if eligible else None,
         "best_B8D_diagnostic":best_focus["variant"],"unchanged_preregistered_gates":GATES,
         "original_test_used":False,"professor_test_used":False,"validation_used_in_fit":False,
         "handmade_used_in_fit":False,"main_changed":False}
    (out/"metadata.json").write_text(json.dumps(metadata,ensure_ascii=False,indent=2),encoding="utf-8")
    lines=["# EXP유016 — 고비율 증강 60개 MLP 및 고정 3조건 통합 실측","",
           "모든 값은 성공한 3 Seed 학습 Artifact와 고정 EXP015 모델의 실제 CSV/NPZ에서 집계함.",
           f"23조건 중 과거 집단 Shortcut Gate 통과: {metadata['legacy_gate_pass']}개; "
           f"이번 개별 클래스 Gate 포함 전체 통과: **{len(eligible)}개**.","",
           "| 조건 | 적격 | 공식 Val | HM 숫자 | HM 문자 | 숫자 1.2 Flip | 문자 0.8 Flip | B8D 최악 | 16클래스 최악 |",
           "|---|---|---:|---:|---:|---:|---:|---:|---:|"]
    for x in ordered:
        lines.append("| "+x["variant"]+" | "+("통과" if x["passes_all_predeclared_gates"] else "탈락")+" | "+
            " | ".join(f"{100*x[k]:.2f}%" for k in
            ("official_val_mean","handmade_digits_mean","handmade_letters_mean",
             "digits_1p2_flip","letters_0p8_flip","B8D_max_class_flip","all16_max_class_flip"))+" |")
    lines+=["","## 사전 선택 기준 판정"]
    if eligible:
        w=eligible[0]
        lines.append(f"- 모든 안전 Gate 통과 후보 중 전체 16클래스 최악 Flip 최소: {w['variant']}.")
        lines.append(f"- 공식 Val {100*w['official_val_mean']:.4f}%; HM 숫자 {100*w['handmade_digits_mean']:.4f}%, 문자 {100*w['handmade_letters_mean']:.4f}%.")
    else:
        lines.append("- 안전 Gate를 전부 통과한 조건 없음. **최종 모델 채택 금지**.")
    lines += [f"- 참고: B8D 여섯 극단 변화 최저 조건 {best_focus['variant']}: {100*best_focus['B8D_max_class_flip']:.2f}% (전체 Gate 통과와 구분).",
              "- Family flip은 참값 예측 Accuracy와 동일하지 않음. Hand-made 반복 개발용 데이터이므로 독립 테스트를 의미하지 않음.",
              "- 0.8~1.2 배율 변형은 픽셀 질량과 보간도 바꾸므로 순수 Size 한 요소만 조작했다는 주장은 지양.",
              "- 기존 Main, 최종 모델, Original Test 및 교수님 Test 유지/미사용.",""]
    (out/"SUMMARY.md").write_text("\n".join(lines),encoding="utf-8")
    print("EXP016_AGGREGATE_SUCCESS",len(ranked),len(eligible),flush=True)
if __name__=="__main__":
    p=argparse.ArgumentParser()
    p.add_argument("--high-inputs",type=Path,required=True)
    p.add_argument("--historical-inputs",type=Path,required=True)
    p.add_argument("--out",type=Path,required=True)
    a=p.parse_args()
    run(a.high_inputs,a.historical_inputs,a.out)
```

```python
%%archive_source 05_model_optimize/MLP/Normal/scripts/exp016_officialval_audit_aggregate.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""Aggregate fixed EXP016 models under Official Validation scale interventions."""
from __future__ import annotations
import argparse,csv,json
from pathlib import Path
from collections import defaultdict
def read(p):
    with Path(p).open("r",encoding="utf-8-sig",newline="") as f:return list(csv.DictReader(f))
def save(p,rows):
    assert rows
    with Path(p).open("w",encoding="utf-8-sig",newline="") as f:
        w=csv.DictWriter(f,fieldnames=list(rows[0]));w.writeheader();w.writerows(rows)
def mean(x):return sum(x)/len(x)
def run(root,out):
    root=Path(root);out=Path(out)
    if out.exists():raise FileExistsError(out)
    out.mkdir(parents=True)
    folders=[root/f"exp016-audit-s{s}" for s in (42,43,44)]
    for p in folders:
        meta=json.loads((p/"manifest.json").read_text(encoding="utf-8"))
        assert meta["original_exp006_com_predictions_verified"]
        assert meta["fine_tuning_or_training_performed"] is False
        assert not meta["official_test_used"] and not meta["professor_test_used"]
    rows=[r for p in folders for r in read(p/"officialval_scale_sensitivity.csv")]
    pairs=[r for p in folders for r in read(p/"officialval_pair_errors.csv")]
    save(out/"officialval_scale_all_seed.csv",rows)
    save(out/"officialval_pair_all_seed.csv",pairs)
    grouped=defaultdict(list)
    for r in rows:
        key=(r["model"],r["scale"],r["group"])
        grouped[key].append(r)
    summary=[]
    for (model,scale,group),v in sorted(grouped.items()):
        assert len(v)==3
        s={"model":model,"scale":scale,"group":group,"n_per_seed":int(v[0]["n"]),"seeds":3}
        for col in ("family_flip_rate","mean_abs_delta_p_letter","mean_signed_delta_p_letter","accuracy"):
            s["mean_"+col]=mean([float(r[col]) for r in v])
            s["max_seed_"+col]=max(float(r[col]) for r in v)
        summary.append(s)
    save(out/"officialval_scale_3seed_summary.csv",summary)
    md=["# EXP016 — OfficialVal 크기 개입 교차 진단",
        "",
        "78개 학습 후 기존 사전 Gate를 통과한 9개 후보와 동결된 CoM 원본을 **고정 가중치**로 재평가하였다.",
        "기존 OfficialVal 19,200장의 원본 이미지를 0.8·0.9·1.0·1.1·1.2배 변형한 후 CoM 정렬.",
        "**Hand-made에 대해 사전 등록한 Gate/순위는 변경하지 않는다.**",
        "OfficialVal은 이미 반복 사용된 개발 Validation이지 새로운 독립 Test가 아니다.",
        "",
        "## 숫자 ×1.2 및 문자 ×0.8의 Family 판단 전환율",
        "",
        "| 모델 | OfficialVal 숫자 1.2x flip | OfficialVal 문자 0.8x flip | 원본 OfficialVal Accuracy |",
        "|---|---:|---:|---:|"]
    lookup={(r["model"],r["group"],float(r["scale"])):r for r in summary}
    names=sorted({r["model"] for r in summary})
    for name in names:
        digits=lookup[name,"digits",1.2]
        letters=lookup[name,"letters",.8]
        original=lookup[name,"all",1.]
        md.append(f"| {name} | {100*digits['mean_family_flip_rate']:.2f}% | "
                  f"{100*letters['mean_family_flip_rate']:.2f}% | "
                  f"{100*original['mean_accuracy']:.3f}% |")
    md+=["","## 집중 혼동쌍 B·8·D·0·F·5의 Family Flip","",
        "| 모델 | 문자/숫자 | 0.8x | 1.2x |",
        "|---|---|---:|---:|"]
    for name in names:
        for label in ("8","B","0","D","5","F"):
            c8=lookup[name,"class_"+label,.8]["mean_family_flip_rate"]
            c12=lookup[name,"class_"+label,1.2]["mean_family_flip_rate"]
            md.append(f"| {name} | {label} | {100*c8:.2f}% | {100*c12:.2f}% |")
    md+=["","### 한계 및 결론",
      "- 이미지 배율 조절은 크기뿐 아니라 획 질량, 두께, 보간·topology도 함께 바꾼다.",
      "- Hand-made에서 Size Shortcut이 완화된 후보라도 OfficialVal에서 동일한 강건성을 보이지 않을 수 있다.",
      "- B/8/D/0/F/5 등 특정 문자 쌍에서 높은 Flip이 남는다면 Shortcut 완전 제거를 주장할 수 없다.",
      "- 어떤 모델도 Test를 평가하거나 final 모델에 교체하지 않았다.",
      ""]
    (out/"SUMMARY.md").write_text("\n".join(md),encoding="utf-8")
    (out/"manifest.json").write_text(json.dumps(
        {"experiment":"EXP016_OFFICIALVAL_AUDIT_AGGREGATE",
         "seeds":[42,43,44],"n_fixed_models":len(names),
         "model_fitting_performed":False,"test_used":False,
         "original_preregistered_criteria_modified":False},indent=2),encoding="utf-8")
    print("EXP016_OFFICIALVAL_AUDIT_AGGREGATE_SUCCESS",len(names),len(summary),flush=True)
if __name__=="__main__":
    p=argparse.ArgumentParser()
    p.add_argument("--inputs",type=Path,required=True)
    p.add_argument("--out",type=Path,required=True)
    a=p.parse_args();run(a.inputs,a.out)
```
